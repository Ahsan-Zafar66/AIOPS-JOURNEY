```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
# storing url of the website in an object
url = "https://www.coingecko.com/"

# getting the response using request
response = requests.get(url)

# checking the status of the code
response.status_code
```

```python
# getting the content of the url  y using beautiful soup
soup = BeautifulSoup(response.content,"html.parser")

# # checking the content
# soup
```

```python
results = soup.find("div",{"class":"coingecko-table"}).find("tbody").find_all("tr")
```

```python
# this code will scrape only the first page of website

#	Coin	Price	1h	24h	7d	24h Volume	Mkt Cap	Last 7 Days

# initiating empy list for the data variables to be scraped
coin = []
price = []
one_h = []
twentyfour_h  = []
seven_day = []
volume = []
market_cap = []

# initiating a loop on the results of page # 1, in each iteration 1 line of the table content will be selected
for result in results:
  try:
    coin.append(result.find("span",{"class":"lg:tw-flex font-bold tw-items-center tw-justify-between"}).get_text().strip())
  except:
    coin.append("n/a")
  try:
    price.append(result.find("span",{"class":"no-wrap"}).get_text().strip())
  except:
    price.append("n/a")
  try:
    one_h.append(result.find("td",{"class":"td-change1h change1h stat-percent text-right col-market"}).get_text().strip())
  except:
    one_h.append("n/a")
  try:
    twentyfour_h.append(result.find("td",{"class":"td-change24h change24h stat-percent text-right col-market"}).get_text().strip())
  except:
    twentyfour_h.append("n/a")
  try:
    seven_day.append(result.find("td",{"class":"td-change7d change7d stat-percent text-right col-market"}).get_text().strip())
  except:
    seven_day.append("n/a")
    
  try:
    volume.append(result.find("td",{"class":"td-liquidity_score lit text-right col-market"}).get_text().strip())
  except:
    volume.append("n/a")
  try:
    market_cap.append(result.find("td",{"class":"td-market_cap cap col-market cap-price text-right"}).get_text().strip())
  except:
    market_cap.append("n/a")

crypto_df = pd.DataFrame({"Coin":coin,"Price":price,"One_Hour":one_h,"TwentyFourHour":twentyfour_h,"SevenDay":seven_day,"Volume":volume,"Market_cap":market_cap})

  
  
```

### The following code will scrape any pages you mention in the for loop

```python
#	Coin	Price	1h	24h	7d	24h Volume	Mkt Cap	Last 7 Days
coin = []
price = []
one_h = []
twentyfour_h  = []
seven_day = []
volume = []
market_cap = []

# Initiating a for loop that will iterate 10 times for 10 pages
for i in range(1,10):

  # url of page to scrape
  url = "https://www.coingecko.com/?page="+str(i)
  
  # getting the response using request
  response = requests.get(url)

  # checking the status of the code
  #response.status_code
  # getting the content of the url  y using beautiful soup
  soup = BeautifulSoup(response.content,"html.parser")
  
  # this code will fetch the data of the table for each page
  # tr stands for table rows, the code will fetch the data from only the table from each page and will fetch all rows of that table
  results = soup.find("div",{"class":"coingecko-table"}).find("tbody").find_all("tr")


  
  # initiating another for loop inside the first for loop, this for loop will iterate line bye line on the tabel cotents of each page
  for result in results:
    try:
      coin.append(result.find("span",{"class":"lg:tw-flex font-bold tw-items-center tw-justify-between"}).get_text().strip())
    except:
      coin.append("n/a")
    try:
      price.append(result.find("span",{"class":"no-wrap"}).get_text().strip())
    except:
      price.append("n/a")
    try:
      one_h.append(result.find("td",{"class":"td-change1h change1h stat-percent text-right col-market"}).get_text().strip())
    except:
      one_h.append("n/a")
    try:
      twentyfour_h.append(result.find("td",{"class":"td-change24h change24h stat-percent text-right col-market"}).get_text().strip())
    except:
      twentyfour_h.append("n/a")
    try:
      seven_day.append(result.find("td",{"class":"td-change7d change7d stat-percent text-right col-market"}).get_text().strip())
    except:
      seven_day.append("n/a")
    
    try:
      volume.append(result.find("td",{"class":"td-liquidity_score lit text-right col-market"}).get_text().strip())
    except:
      volume.append("n/a")
    try:
      market_cap.append(result.find("td",{"class":"td-market_cap cap col-market cap-price text-right"}).get_text().strip())
    except:
      market_cap.append("n/a")

# storing all the data of 10 pages in a dataframe
crypto_df2 = pd.DataFrame({"Coin":coin,"Price":price,"One_Hour":one_h,"TwentyFourHour":twentyfour_h,"SevenDay":seven_day,"Volume":volume,"Market_cap":market_cap})

  
  
```
