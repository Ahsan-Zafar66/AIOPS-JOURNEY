```python
# importing the libraries
import pandas as pd
import numpy as np
from scipy.stats import f_oneway
from statsmodels.stats.multicomp import pairwise_tukeyhsd
from statsmodels.stats.multicomp import MultiComparison
```

```python
# Sample data
data = {'Class': ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C'],
        'Score': [85, 88, 82, 78, 75, 80, 92, 88, 94]}
df = pd.DataFrame(data)
```

```python
df
```

```python
# One-way ANOVA test
grouped_data = [group['Score'] for name, group in df.groupby('Class')]
```

```python
grouped_data
```

```python
f_statistic, p_value = f_oneway(*grouped_data)

print("One-way ANOVA results:")
print("F-statistic:", f_statistic)
print("p-value:", p_value)
```

```python


# Post hoc analysis - Tukey's HSD test
multicomp = MultiComparison(df['Score'], df['Class'])
result = multicomp.tukeyhsd()

print("\nPost hoc analysis - Tukey's HSD test results:")
print(result)

```

# Bonferroni POST HOC ANALYSIS

```python
# pairwise tukeyhsd performs the boneferroni test
from statsmodels.stats.multicomp import pairwise_tukeyhsd
```

```python
# Sample data
data = {'Class': ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C'],
        'Score': [85, 88, 82, 78, 75, 80, 92, 88, 94]}
df = pd.DataFrame(data)

# One-way ANOVA test
grouped_data = [group['Score'] for name, group in df.groupby('Class')]
f_statistic, p_value = f_oneway(*grouped_data)

print("One-way ANOVA results:")
print("F-statistic:", f_statistic)
print("p-value:", p_value)

# Post hoc analysis - Bonferroni correction
result = pairwise_tukeyhsd(df['Score'], df['Class'], alpha=0.05)

print("\nPost hoc analysis - Bonferroni correction results:")
print(result)

```

# CHI-SQUARE TEST


```python
import pandas as pd
from scipy.stats import chi2_contingency

# Sample data
data = {'Gender': ['Male', 'Male', 'Female', 'Female', 'Male', 'Female'],
        'HeartDisease': [1, 1, 0, 1, 0, 1]}
df = pd.DataFrame(data)

# Contingency table
contingency_table = pd.crosstab(df['Gender'], df['HeartDisease'])

# Chi-Square test
chi2, p_value, _, _ = chi2_contingency(contingency_table)

# Number of observations
n = df.shape[0]

# Cramer's V calculation
phi = np.sqrt(chi2 / n)

print("Cramer's V:", phi)
```

# Leven's Test of HOMOGENEITY
- The null hypothesis of Levene's test is that the variances of the groups are equal, while the alternative hypothesis is that at least one group has a different variance. If the p-value from Levene's test is less than the chosen significance level (e.g., 0.05), we reject the null hypothesis, indicating that the variances are not equal among the groups

```python
import numpy as np
from scipy.stats import levene

# Sample data
group_a = [12, 14, 10, 15, 18]
group_b = [9, 11, 10, 12, 8]
group_c = [16, 13, 15, 17, 20]

# Perform Levene's test
statistic, p_value = levene(group_a, group_b, group_c)

print("Levene's test results:")
print("Test statistic:", statistic)
print("p-value:", p_value)
```
