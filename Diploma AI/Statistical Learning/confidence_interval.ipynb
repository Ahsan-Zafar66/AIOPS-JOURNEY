# Confidence Interval
- The notebook contains code to calculate confidence interval manuallya s well as using the scipy library

# Question
- What is the average bill payed by ethe customers? Build a confidence interval of 95% around the average bill.

```python
# importing the libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
# loading the tips dataset
tips = sns.load_dataset("tips")
```

```python
# checking the data
tips.head()
```

# Manual Method Formula
# MEAN +- (z_score value * Standard Error)



```python
# calculating average bill
avg_bill = tips["total_bill"].mean()

# calculating std
std = tips["total_bill"].std()

# calculating standard error
se = std/np.sqrt(len(tips) - 1 )
```

```python
# 95% confidence interval

# z-value got from z-table for value 0.975 which is 1.96, because we want the middle 95% so that each tail has 2.5% remaining, so we will take zscore value for 97.5%
z_value = 1.96

# calculating upper bound using formula
upper_bound = avg_bill + (z_value * se)

# calculating lower bound using formula
lower_bound = avg_bill - (z_value * se)
```

```python
# 95% confidence interval usig manual method
print("95% confidence interval for avg bill is {} - {} ".format(lower_bound,upper_bound))
```

```python
# calculating standard error using scipy
std_error = stats.sem(tips["total_bill"])


# degrees of freedon
degrees_of_freedom = len(tips) - 1

# setting alpha value
alpha = 0.05

# calculating critical value
t_critical = stats.t.ppf(1 - alpha / 2, degrees_of_freedom)

# Calculate the lower and upper bounds of the confidence interval
lower_bound = avg_bill - t_critical * std_error
upper_bound = avg_bill + t_critical * std_error

# printing the 95% confidence interval
print(f"95% Confidence Interval: ({lower_bound:.3f}, {upper_bound:.3f})")
```

```python
# 95% confidence interval usig manual method
print("95% confidence interval for avg bill is {} - {} with manual method ".format(lower_bound,upper_bound))

# printing the 95% confidence interval
print(f"95% Confidence Interval: ({lower_bound:.3f}, {upper_bound:.3f})  with scipy library")
```
