# MANOVA
- MANOVA (Multivariate Analysis of Variance) is a statistical technique used to analyze the differences between groups of multivariate data.
- Here whe have more than one category or groups and more than one variable to study

```python
# importing libraries
import numpy as np
import pandas as pd
import statsmodels.api as sm
import seaborn as sns
from statsmodels.multivariate.manova import MANOVA
```

```python
# Load the iris dataset
iris = sns.load_dataset("iris")

# Separate the dependent and independent variables
X = iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
y = iris['species']

```

```python
# Perform MANOVA
maov = MANOVA.from_formula('sepal_length + sepal_width + petal_length + petal_width ~ species', data=iris)
print(maov.mv_test())

```
