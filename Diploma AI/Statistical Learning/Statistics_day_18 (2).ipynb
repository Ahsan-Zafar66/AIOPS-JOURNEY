- ANOVA
- 3+ levels or groups

- Two-way ANOVA, For example, examining the height and weight of children, young and adult. Groups being studied are 3 but two variables are being studied which are height and weight

- Repeated measures of ANOVA ( 3+ paired groups, scale up of paired t-test). We can use paired t-test for femal math and female stats marks, but what if we want to compare math, stat and biology? Here comes repeated measures of ANOVA which allows 3+ paired gorups, by using it we can analyze female math, female stat and female biology.

# Research Design

# CRD
- Completely Randomized Design
- In CRD, each experimental unit is randomly assigned to one of the treatment groups.

- For example we need to test new food effect on sheeps. We have 20 shpees, we will randomly sellect 10 sheep in each group. We will give food 1 to group # 1 and food 2 to group number 2.

- Mtlb ky ye khoraak sabky liye achi hai ya achi nhi hai.


# RCBD
- Randomized complete block design
- In this design we form a block and than follow CRD.
- For example we have 20 sheep, we will make blocks of male sheep and female sheep. We will randomly select male sheeps and form two groups, one group will recieve food1 and the other one food2. Repeat the proces for female sheeps as well.
- RCBD helps when we have multiple variable.

- mtlb ye khorak bakriyon ya bakron ky liye achi hai

# LSD
- latin Square design
- Similar to square
- Rows = Columns, e.g 2 rows 2 columns, 3 rows 3 columns in square fashion
- Eac experimental unit (a single sheep) will get all the treatments but in square style

|----|Week # 1|Week # 2|Week # 3|
| --- | --- | --- |-----|
| Sheep #1|Food-1 | Food-2 | Food-3 |
| Sheep #2| Food-3 | Food-1 | Food-2 |
| Sheep #3|, Food-2 | Food-3| Food-1 |


- No repetition of treatment in rows or columns.
- Useful when you have less resources.

# Correlation Test
### 1-Pearson's Correlation Coefficient
- Test whether two samples have a linear relationship
### Assumptions 
1) Observations in each sample are independent and identically distributed (iid)
2) Observations in each sample are normally distributed 
3) Observations in each sample have the same variance
4) Interpretation

H0 : The two samples are independent
H1 : There is a dependency between the samples

```python
# Example of a pearsonr correlation test
from scipy.stats import pearsonr
stat,p = pearsonr(titanic['age'],titanic["fare"])
if p > 0.05:
    print("Probably independent")
else:
    print("probably dependent")
```

```python
# Example of a spearman correlation test
from scipy.stats import spearmanr
stat,p = spearmanr(titanic['age'],titanic["fare"])
if p > 0.05:
    print("Probably independent (No Correlation)")
else:
    print("probably dependent (Correlation Exists)")

print("stats=%.3f,p = %.3f"%(stat,p))
```

### 2-Spearman Correlation (non-paramteric distribution (mtlb ky distribution normal na ho to bhi laga skty hain))
Tests whether two samples have a monotonic relationship
### Assumptions
1) Observations in each sample are independent and identically distributed (iid)
2) Observations in each sample can be ranked 
3) Interpretation

H0 : the two sample are independent
H1 : There is a dependency between the samples

```python
# dropping null values from the dataset
from scipy.stats import spearmanr
titanic3 = titanic.dropna()
stat,p = spearmanr(titanic3['age'],titanic3["fare"])

if p > 0.05:
    print("Probably independent (No Correlation)")
else:
    print("probably dependent (Correlation Exists)")

print("stats=%.3f,p = %.3f"%(stat,p))
```

# Chi-square Test
Tests whether two categoricaal variables are related or independent.

### Assumptions 
Observations used in the calculation of the contingency table are independent. 25 or more examples in each cell of the contingency table.
Interpretation:
- H0 : The two samples are independent
- H1 : There is a dependency between the samples

[Chi-square test](https://www.youtube.com/watch?v=f53nXHoMXx4)

```python
# Example of chi=-square test
from scipy.stats import chi2_contingency
table = [[20,30,30],[100,110,90]]
stat,p,dof,expected = chi2_contingency(table)
if p > 0.05:
    print("Probably independent")
else:
    print("probably dependent")

print(p)
```

# 3-Parametric Statistical Hypothesis Test

Test whether the means of two independent samples are significantlly different

### Assumptions
1) Observations in each sample are independent and identically distributed (iid)
2) Observations in each sample are normally distributed 
3) Observations in each sample have the same variance

### Interpretation

H0 : The means of the two sample are equal
H1 : The means of the two sample are not equal


```python
# Example of Student's t test
from scipy.stats import ttest_ind
#data1 = [0.837,2.817,0.121,-0.945,-0.055,-1.436]
#data2 = [1.142,2.424,-0.55,1.212,3.21]
without_drug = [1.2]
withdrug = [1.05]

stat,p = ttest_ind(without_drug,withdrug)
if p > 0.05:
    print("Probably the same distribution")
else:
    print("Probably different distribution")
```

# Paired student t-test (larkiyon ky math mn kitny marks hain or larkiyon ky statistics mn kitny marks hain, same sample sy 2 kisam ki cheezein nikal rhy hon)
Test whether the means of two paired samples are significantly different 

### Assumptions

1) Observations in each sample are independent and identically distributed (iid)
2) Observation in each sample are normally distributed
3) Observation in each sample have the same variance
4) Interpretation

### Interpretation
H0 : The means of the sample are equal
H1 : The means of the sample are not equal


```python
# Example of Paired t test
from scipy.stats import ttest_rel
data1 = [0.837,2.817,0.121,-0.945,-0.055,-1.436]
data2 = [1.142,2.424,-0.55,1.212,3.21,1.21]
stat,p = ttest_rel(data1,data2)
if p > 0.05:
    print("Probably the same distribution")
else:
    print("Probably different distribution")
```

# Analysis Of Variance Test (ANOVA) (3 ya 3 sy zyada ka hota hai)
Test whether the means of two or more independent samples are significantly different
### Assumptions

1) Observations in each sample are independent and identically distributed (iid)
2) Observation in each sample are normally distributed
3) Observation in each sample have the same variance

### Interpretation

H0 : The means of the sample are equal

H1 : one or more of the means of the samples are unequal


 

```python
# Example of ANOVA
from scipy.stats import f_oneway
data1 = [0.837,2.817,0.121,-0.945,-0.055,-1.436]
data2 = [1.142,2.424,-0.55,1.212,3.21,1.21]
data3 = [1.142,2.424,-0.55,-0.945,-0.055,-1.436]
stat,p = f_oneway(data1,data2,data3)
if p > 0.05:
    print("Probably the same distribution")
else:
    print("Probably different distribution")
```

```python
# Example of ANOVA
from scipy.stats import f_oneway
data1 = [3,2,1]
data2 = [5,3,4]
data3 = [5,6,7]
stat,p = f_oneway(data1,data2,data3)
if p > 0.05:
    print("Probably the same distribution")
else:
    print("Probably different distribution")
```
