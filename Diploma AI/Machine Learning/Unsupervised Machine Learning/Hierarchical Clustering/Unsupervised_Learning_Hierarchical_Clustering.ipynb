# Hierarchical Clustering
- Hierarchical clustering is a method used in unsupervised machine learning to group similar data points into clusters based on their distances or similarities. It organizes data into a tree-like hierarchical structure known as a dendrogram.

### How Hierarchical Clustering Works:

- **Distance Measurement:**
- Hierarchical clustering starts by treating each data point as a single cluster.
- It calculates the distances or similarities between all pairs of data points. Common distance metrics include Euclidean distance, Manhattan distance, or correlation-based distances.

- **Merge or Divide:**
- Based on these distances, the algorithm progressively merges or divides the closest or most similar data points or clusters into larger or smaller clusters.

- **Dendrogram Construction:**
- As the process continues, a dendrogram is constructed, visually representing the hierarchy of clusters. The y-axis represents the distance or dissimilarity at which clusters are merged.

- **Cutting the Dendrogram:**
- Decisions about the number of clusters can be made by cutting the dendrogram at a particular height, dividing it into the desired number of clusters.

```python
# importing libraries
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
# Generating sample data
X, _ = make_blobs(n_samples=300, centers=2, random_state=42)
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1])
plt.show()
```

```python
# Performing hierarchical clustering
model = AgglomerativeClustering(n_clusters=2, linkage='ward')
model.fit(X)
```

```python
# Plotting the dendrogram
linked = linkage(X, 'ward')
plt.figure(figsize=(8, 6))
dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=True)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Sample Index')
plt.ylabel('Distance')
plt.show()
```

```python
# Plotting the dendrogram
linked = linkage(X, 'ward')
plt.figure(figsize=(8, 6))
dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=True)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Sample Index')
plt.ylabel('Distance')
plt.axhline(y=80, color='red', linestyle='--', label='Horizontal Line at y=0.5')
plt.show()

```

```python
# Visualization of clustered data
plt.scatter(X[:, 0], X[:, 1], c=model.labels_, cmap='viridis')
plt.title('Clustered Data')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```
