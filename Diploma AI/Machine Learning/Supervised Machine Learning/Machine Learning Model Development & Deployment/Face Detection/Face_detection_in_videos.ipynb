```python
import cv2
import numpy as np
```

```python
# loading the harcascade model
har = cv2.CascadeClassifier("D:\AIOPS\Diploma AI\Machine Learning\Machine Learning Model Development & Deployment\haarcascade_frontalface_default.xml")
```

```python
def face_detect(img):

    # converting the image into grayscale
    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    faces = har.detectMultiScale(gray_img,scaleFactor=1.1,minNeighbors=5)

    for x,y,w,h in faces:
        cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,255),2)

    return img
```

```python
# give the video path or write 0 to open webcam
cap = cv2.VideoCapture(0)
```

```python
while True:
    status, frame = cap.read()
    if status == False:
        break
    frame = face_detect(frame)
    cv2.imshow("detect",frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break 

cv2.destroyAllWindows()
cap.release()
```
