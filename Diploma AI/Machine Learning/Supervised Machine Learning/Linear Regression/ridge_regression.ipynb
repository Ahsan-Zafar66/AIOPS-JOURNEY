# Ridge Regression:

- Ridge Regression, also known as Tikhonov regularization or L2 regularization, is a linear regression technique used to prevent overfitting by adding a regularization term to the linear regression cost function. It works by introducing a penalty on the magnitudes of the coefficients, which helps in controlling the complexity of the model and reduces the impact of collinearity among predictor variables.

# Working Principle:

- In Ridge Regression, the cost function is modified to include the sum of squared coefficients multiplied by a regularization parameter (λ or alpha). The goal is to find coefficient values that not only fit the data well but also keep them small, thus reducing the model's complexity.

# Parameters:

alpha (λ): The regularization parameter controls the strength of the penalty. Higher values of alpha result in smaller coefficient values. The choice of alpha requires tuning and is usually determined through techniques like cross-validation.
Regularization Formula in Ridge Regression:

# The cost function for Ridge Regression is:
- Cost = Least Squares Loss + alpha * (sum of squared coefficients)

- In Ridge Regression, the regularization parameter alpha (λ) controls the strength of the penalty applied to the coefficients of the linear regression model. The effect of alpha on Ridge Regression can be understood as follows:

# Small Alpha (Weak Regularization):
- When alpha is very small (close to zero), the penalty term becomes negligible, and Ridge Regression becomes similar to ordinary linear regression. The model's coefficients are allowed to grow freely, and it may result in overfitting, especially when dealing with multicollinearity.

# Intermediate Alpha:
- As you increase alpha to an intermediate value, the penalty term starts to have a noticeable impact on the coefficients. The model's complexity is reduced, and the coefficients are shrunk towards zero. This helps in controlling overfitting and improving the model's generalization to new data.

# Large Alpha (Strong Regularization):
- When alpha is very large, the penalty term dominates the cost function. This leads to significant shrinkage of the coefficients towards zero. The model becomes simpler, and it's less likely to overfit the training data. However, very large alpha values may lead to underfitting if the coefficients are excessively shrunk.

# Choosing the Right Alpha:

- The choice of alpha depends on the trade-off between model simplicity (bias) and its ability to fit the training data (variance). Cross-validation is commonly used to find the optimal alpha value. You can try a range of alpha values and select the one that provides the best performance on validation data. Techniques like GridSearchCV in scikit-learn can help automate this process.

# In general:

- If your model is overfitting, consider increasing alpha.
- If your model is underfitting, consider decreasing alpha.
- It's important to note that the optimal alpha value can vary based on the specific dataset and problem at hand. Cross-validation helps in finding the right balance between reducing overfitting and preserving model performance.








```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import numpy as np
from sklearn.linear_model import LinearRegression
import pandas as pd
```

```python
df = pd.read_csv("/content/sample_data/california_housing_train.csv")
```

```python
df.head()
```

```python
X = df.drop("median_house_value",axis=1)
y = df[["median_house_value"]]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=21)
```

```python
# linear regression
lr_model = LinearRegression()

# Fit the model to the training data
lr_model.fit(X_train, y_train)

# Predict on test data
y_pred = lr_model.predict(X_test)

# rmse on training
y_pred_train = lr_model.predict(X_train)
rmse_train = np.sqrt(mean_squared_error(y_train,y_pred_train))

# Calculate Mean Squared Error (MSE)
mse_test = mean_squared_error(y_test, y_pred)
rmse_test = np.sqrt(mse_test)
#print("Linear Regression Coefficients:", lr_model.coef_)
print("Root Mean Squared Error Testing data:", rmse_test)
print("Root Mean Squared Error on training data:",rmse_train)
```

```python
# Create a Ridge Regression model with a specified alpha (λ)
alpha = 10
# You can tune this value
ridge_model = Ridge(alpha=alpha)

# Fit the model to the training data
ridge_model.fit(X_train, y_train)

# Predict on test data
y_pred = ridge_model.predict(X_test)

# training data prediction
y_pred_train = ridge_model.predict(X_train)
rmse_train = np.sqrt(mean_squared_error(y_train,y_pred_train))

# Calculate Mean Squared Error (MSE)
mse_test = mean_squared_error(y_test, y_pred)
rmse_test = np.sqrt(mse_test)
#print("Ridge Regression Coefficients:", ridge_model.coef_)
print("Root Mean Squared Error:", rmse_test)
print("Root Mean Squared Error:", rmse_train)
```
