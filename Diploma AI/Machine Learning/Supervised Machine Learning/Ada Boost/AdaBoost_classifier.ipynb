# AdaBoost
- AdaBoost (Adaptive Boosting) is an ensemble learning method that is used for both classification and regression tasks. It belongs to the family of boosting algorithms, which combine multiple weak learners (classifiers/regressors) to create a single strong learner. The main idea behind AdaBoost is to focus on the data points that are misclassified by the previous base learners, assigning them more weight, and training new base learners to improve their classification.

- Following is a detailed explanation of how AdaBoost works:

- **Data Weighting:** AdaBoost assigns equal weight to all data points in the training dataset initially.

- **Base Learner Training:** AdaBoost trains a base learner (e.g., decision tree, stump) on the training data. The base learner's task is to classify the data points.

- **Weighted Error Calculation:** AdaBoost calculates the weighted error rate of the base learner, which is the sum of the weights of misclassified points divided by the total weight. It gives more weight to data points that the base learner misclassifies.

- **Base Learner Weight:** AdaBoost calculates a weight for the base learner based on its performance. Better-performing base learners are assigned higher weights.

- **Data Point Weight Update:** AdaBoost increases the weight of the misclassified data points and decreases the weight of correctly classified data points. This makes the previously misclassified data more important for the next round.

- **Repeat:** Steps 2-5 are repeated for a fixed number of iterations or until a predefined level of accuracy is achieved.

- **Final Classification:** AdaBoost combines the individual base learners to form a strong classifier. The base learners with higher weights have more influence on the final classification.

- AdaBoost can be applied to various types of base learners, but weak learners are commonly used, as they tend to be simple and have good generalization performance. The key idea is that by iteratively training and combining these base learners, AdaBoost focuses on the examples that are difficult to classify, ultimately creating a strong and accurate classifier.

# Hyperparameters of AdaBoost:

- Base Estimator: You can specify the base estimator, which is typically a decision tree. You can choose the depth of the trees, which controls their complexity.

- Number of Estimators (n_estimators): This hyperparameter controls the number of base learners in the ensemble. A larger number of estimators can increase the model's accuracy, but it may also lead to longer training times.

- Learning Rate (learning_rate): The learning rate shrinks the contribution of each base learner to the final prediction. A smaller learning rate requires more estimators for the same level of accuracy.

# Advantages of AdaBoost:

- It is a powerful ensemble method that can significantly improve the accuracy of weak base learners.
- It adapts to the data by giving more importance to previously misclassified samples.
- It is less prone to overfitting compared to other complex models.
Disadvantages of AdaBoost:

- It can be sensitive to noisy data and outliers.
0 The choice of a weak learner and the number of estimators can impact the model's performance.
- Training time may be longer if the number of estimators is high.

```python
import seaborn as sns
from sklearn.ensemble import AdaBoostClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import warnings
warnings.filterwarnings("ignore")
```

```python
# Load the Iris dataset
iris = sns.load_dataset("iris")
X = iris.drop("species",axis=1)
y = iris[["species"]]
```

```python
X.head()
```

```python
y.head()
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Create an AdaBoost classifier
adaboost = AdaBoostClassifier(n_estimators=50, random_state=42)

# Train the classifier
adaboost.fit(X_train, y_train)

# Make predictions
y_pred = adaboost.predict(X_test)
```

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
```

```python
# Generate a classification report
report = classification_report(y_test, y_pred,)
print(report)
```

```python
# Create a confusion matrix
matrix = confusion_matrix(y_test, y_pred)
print(matrix)
```
