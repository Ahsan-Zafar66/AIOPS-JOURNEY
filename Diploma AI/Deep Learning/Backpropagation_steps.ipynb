# Backpropagation Steps
- 1) You initiate weights and biases randomly
- 2) You loop through your dataset the number of times equal to your number of data points or rows.
- 3) You would do forward propagation for each row
- 4) You will now calculate loss for the data point
- 5) Now, on the basis of loss fuctions you will update your weights and biases using Gradient descent. The formula for updating wights or biases is
formual: **weight_new = weight_old - learning_rate(partial derivative of loss/partial derivative of weight old)**. 
- 6) Now you will repeat from step 3 to 5 for all observaions or rows one by one. and calculate the average loss for this epoch

- Now after completing on the whole dataset, you repeat from the begining for all data points again. This looping is called epochs.


# Learning rate
- Learning rate is used to smooth the steps of graident descent. It's value used generally is 0.1 or 0.01.

# What is convergence?
- The convergence means when the slope of loss with respect to trainable parameters approaches to 0, or when the new weight equals the old weight and not ver difference in them. When it occurs gradient descent have reached minima.

# Main Formula
- New_weight = Old_weight - learning rate * (partial derivative of Loss with respect to the weight or bias)

- If the derivative of loss with respect to the weight or bias is positive, this means when weight or bias increase loss increases, so we should reduce the weight or bias value, that is why negative sign is used.

- If the deriative of loss with respect to the weight or bias is negative, this means by increasing the weight or bias loss decrease, so we should increase the weight or bias, the slope will be negative and negative signe multiplied by negative becomes positive and the value of weight or bias will be increase.

- The main aim of gradient descent is to reach the minima where the slope of loss with respect to weight and bias is 0, at this zero the values of weight or bias we need to make the loss minimum. We need to search for weight and bias values that make the loss zero.
