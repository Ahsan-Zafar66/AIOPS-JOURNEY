# Activation Functions
- Activation functions are a critical part of the design of a neural network
- If the output range of the activation function is limited, than it may be called a squashing function.
- Activation functions are mathematical equations that determine the output of a neural network model. They help the network to use the important information and suppress the noise
- In a neural network activation functions are utilized to bring non-linearities into the decision boundary. For example an image of a cat and a dog are very different. There is no relationship between them, so they are non-linear.
- The goal of introducing non-linearities into data is to simulate a real world situation.


# Perceptron

# Ideal Properties of Activation Functions
- Non-linear: all data in the world is not linear or linearly separable. There are non-linear regression examples and classification as well, so to deal with this kind of data we have non-linear activation functions
- Diferentiable: The activation function should be differentiable so gradient descent could work on it.
- Computationally inexpensive: It should be computationally inexpensive, fast
- Zero Centered: It should be zero centered or in other words normalized, mean = 0. For example Tanh.
- Non-Saturating: It should not squish the inputs, like sigmoid squishes the inputs whatever the input is you get a value between 0 and 1, for tanh it squishes the value from -1 to 1. For example ReLu, formula max(0,x). If the activation function is saturating like it squishes the inputs than it has the problem of vanishing gradient

# Linear Activation Function
- The linear activation function is also known as no activation and or identity function, where the activation is proportional to input
- The function does not do anything to the weighted sum of the input, it simply spits out the value in the neuron calculated due to the inputs.
- Formula  f(x) = x, apply a function on x you will get x again.

# Problems with linear activation function
- THere are two major problems with linear activation function:
- It is not possible to use backpropagation as the derivative of the function is a constant and has no relation to the input x.
- All layers of the neural network will colapse into one if a linear activation function is used. No matter the number of layers in the neural network, the last layer will still be a linear function of the first layer.
- So, in short the linear function turns the neural network into just one layer.



# Non-Linear activation function
- The  limited power linear activation function (simply a linear regression model), this does not allow the model to create complex mappings between inputs and output (as linear regression is not able to capture non-linear relation ship).

# Limitation of Linear functions solved by non-linear one
- They allow the backpropagation because now the derivative function would be related to the input, and it is possible to go back and understand which weights in the input neurons can provide a better predction
- They allow the stacking of multiple layers of neurons as the output would be now non-linear combination of input passed through multiple layers. Any output can be represented as a functional computation in a neural network


# Binary Step Function
- Binary step function is a threshold based activation function whcih means:
- After a certain threshold neuron is activated
- below the said threshold neuron is deactivated
- It can be used for binary classification probelem as the name suggests, However it cannot be used for classifying multiple classes.

# Sigmoid Function
- Smooth curve between o and 1
- The function take any real values as inputs and outputs in a range of 0 to 1
- The larger the input (positive) the closer the output value will be 1 whereas smaller the input (more negative) the closer the value to 0

- Formula:
- f(x) =  1/ 1 + e ^ (-x), where e is the base of natural logarithm

# Advantages
- Sigmoid function, it is widely used for models where we want to predict probabilities as an output.
- It normalizes the output of each neuron
- The function is differentiable and provides a smooth gradient  i-e preventing jumps in output values (like binary step function). This is represented by the S shape of of the sigmoid activation function

# Disadvantages
- However, sigmoid function makes almost no change in the prediction of very high or low inputs which ultimatly results in neural network refusing to learn further, this problem is know as the vanishing gradien. - Computationally expensive

# Tanh Function (Hyperbolic Tangent)
- Very similar to the sigmoid/logistic activation function
- Even has the same S shape
- **But the difference is output range of -1 to 1**
- The larger the output value closer will be to 1 the lower the output value closer will be to -1

- Formula:
- f(x) = (e^x - e^x) /  (e^x + e^x), e is base of natural logarithm

# Advantages:
- Zero centered means telling even for negative values
- The prediction is simple based on a threshold values

# Disadvantages
- It also has vanishing gradient problem

# ReLU Function (Rectified Linear Unit)
- In this function, output for the postive inputs can range from 0 to infinity
- Although it gives impression of Linear Function, but it allows backpropagation while making it efficiently
- The main catch is that the ReLu function does not activate all the neurons at the same time
- The neurons will only be deactivated if the output of the linear transformation is less than 0

- Formula:
- f(x) = Max(0,x)

# Advantages
- Since only a number of neurons are activated, the ReLU function is far more computationally efficient when compared to the sigmoid and tanh function. Sirf unhi neurons ko thk krta hai jinhein krny ki zaroorat hai sabko thk krny nhi beth jata.
- ReLU accelerates the convergence of gradient descent towards the global minimum of the loss function due to it's linear and non-saturating property.

# Disadvantages
- When the input is zero or a negative values, the function outputs zero and it hinders with the back propagation. This problem is known as the dying ReLU problem. - It cannot be differentiable
- Not zero centered, we use batch normalization, it normalized the output of a hidden layer and than pass these normalized data into the next layer

# Reason For Dying ReLu
- High Learning Rate, the more the learning rate the more will be the negative value to be subtracted from old weight while figuring out the new weight and ther term will become negative, and the neurons will be dead as their output will be zero.
- High Negative Bias (w1x1 + w2x2 + b), the more negative biase more this term will be negative.

# Solutions for dying ReLu
- Set Low learning rate
- Set bais equal to 0.01, it is proven through studies

# Leaky ReLU
- Leaky ReLU is the most common and effective method to solve the dying ReLU problem
- It is nothing but an imporved version of the ReLU function
- It adds a slight slope in the negative range to prevent the dying ReLU issue.

- formula
- f(x) = max(x * 0.1, x)

# Advantages
- Same as that of ReLU
- It enables backropagation for negative values as well
- By making this minor modification for negative input values the gradient of the left side of the graph comes out to be a non-zero value. Therefore, we would no longer encounter dead neurons in that region

# Disadvantage
- only a small portion of left side of graph is covered

# Parametric ReLU
- It aims to solve the problem of gradient becoming zero for left half of the axis
- this function provides the slope of the negative part of the function as an argument a. By performing Backpropagation, the most appropriate value for a is learnt.
- formula:
- f(x) = max(ax,x), where a is slope parameter for negative values

- **In leaky ReLU you specify a value for slope 0.1 but for parametric it learns itself**

# Advantages
- The parametric ReLu function is used when rhe leaky ReLU function still fails at solving the problem of dead neurons , and the relevant information is not successfully passed to the next layer

# Disadvantages
- The function limitation is that it may perfrom differentl for different problems based on the value of slope parameter a.

# ELU (exponential linear unit)
- x >= 0, 1
- x < 0, a(e^x - 1)

# Advantages
- Close to zero centered, due to this convergence to solution is faster
- Experiments show generalized results are better
- Dying relu does not exist as some value exist for gradients even for values less than 0
- Continously Differentiable

# Disadvantage
- Expensive Computationaly

# Selu
- It is same as Elu just the output is multiplied by a lambda value
- The value for a is 1.6732 and for lambda is 1.0507, these are fixed values not trainable.

# Advantage

# Use of each ReLU
- If you have non-negative values use simple ReLU
- if you have only a tiny fraction of negative values use Leaky ReLU.
- If you have a lot of negative values like 50% than use parametric ReLU.


# Softmax Function
- The softmax function is generalized form of sigmoid function for multiple classes.
- It is  amathematical function that converts vector of numbers into a vector of probabilities
- The output is interpreted as probability for each class
- The softmax function is described as combinaton of multiple sigmoid functions.
- It is most commonly used a s a activation function in the last layer of multi-class classification problem

# Advantages
- It can be used for multiclass classification
- It normalizes the output for each class between 0 and 1, and divides by their sum gving the probability of the input value being in a specified class
- For neural netwroks that needs to be categorized into numerous categories. Softmax is often employed exclusively for the output layer.


# How to chose an activation function
- Use sigmoid function in the output layer and tanh on all other places (hidden layers). But both have vanishing gradient problem

- Use ReLU for hidden layes, it is more efficeint and it's math is simple, however it also has vanishing gradient problem. You can use Leaky ReLU and parametric ReLU as well.

# Hidden Layer
- A hidden layer is a layer in a neural network that recieve input from another layer (it can be input or it can be another hidden layer) and provide output to another layer (it can be another hidden layer or an output layer)

- A hidden layer does not directly contact the input data or produce outputs for a model at least in general
- A neural network may have zero or more hidden layers
- Typically, a differentiable non-linear function is used in the hidden layers of a neural network. This allows the model to learn more complext functions than a network trained using a linear activation function.

- Most widely used activation function for hidden layers are
- ReLU, most used, less vanishing gradient
- Sigmoid, most used on second number, high vanishing gradient
- Tanh, third number, high vanishing gradient
- The Modern Neural network  with common architecture, such as MLP and CNN, will make use of the ReLU activation function or extensions
- Recurrent Neural Network still use commonly Tanh or sigmoid function or both. For example LSTM commonly uses the sigmoid function for recurrent connections and the Tanh for output

- MLP use ReLU
- CNN use ReLU
- Recurrent use Tanh or sigmoid
- If you are not sure apply different functions and used the best one


# Can we use different activation function in hidden layers?
