# Loss Function
- Loss function is a method of evaluting how well your algorithm is modeling your dataset.

# Loss Functions in deep learning

# Regression

# Means Squared Error (MSE)
- Mean Squared Error:
- Not Robust to outliers, effected by outliers greatly as square of quantity is taken.
- it is differentiable and Gradient Descent can be applied.
- use when no outliers in the data

# Mean Absolute Error (MAE)
- Mean Absolute Error:
- Robust to outliers, effected less by outliers as no square of quantity is taken
- It is a mod and it is nt differentiable gradient descent cannot be applied directly rather sub gradients are calculated which are tricky
- Use when outliers in data

# Huber Loss
- Huber Loss, Huber loss function is a type of loss function used in deep learning for regression problems. It is a combination of Mean Squared Error (MSE) and Mean Absolute Error (MAE) loss functions, and it is less sensitive to outliers than MSE.
- It works very well when there are lot of outliers. 
- The Huber loss function uses a hyperparameter called delta to determine the threshold for when to switch from quadratic (MSE, mean squared error) to linear (Mean Absolute Error or MAE) loss. When the absolute difference between the true value and predicted value is less than or equal to delta, the loss function uses the quadratic loss, and when the absolute difference is greater than delta, it uses the linear loss.
