# Gradient Descent
- Gradient descent is a popular algorithm to perform optimization and by far the most common way to optimize neural networks.

# Batch Gradient Descent (or Vanila Gradient Descent)
- in batch gradient descent, we do not calculate loss for a single data point than update the weights and move to the next data point until we are done on all the data points, in batch gradient descent we update the weights after we have calculated the loss on the entire data, we will get prediction for first data point, than second till the last data point, at the end of a epoch we update the weights. Normally, we update weights after calculating loss for each data point but in batch gradient descent we update weights after an epoch completes.

- **-  No. of times the weights are updated will be equal to number of epochs.**
- For example, we have 50 data points in our dataset and number of epochs are 10, 10 times the weights and biases will be updated.
- To use batch gradient descent, use batch_size in fit command and it should be equal to number of rows in your data

# Stochastic Gradient Descent
- It is normal gradient descent we have studied, weights are updated for each data point and update the weights, move to next data point again update the weights until all data points are done, the same steps are repeated the number of times epochs is chosen. At the end of a epoch we calculate the average loss for all the data points, and that is the loss for that epoch.

- For example, we have 50 data points in our dataset and number of epochs are 10, 500 times the weights and biases will be updated.

- To use stochastic gradient descent, use batch_size in fit command and it should be equal to 1.


# Which one is faster?
- batch gradient descent will be faster as the number of updates equal to number of epochs

# Which will converge faster towards the solution?
- Stochastic Gradient Descent will converge towards the solution faster, as a large number of updates will be made.

# Batch Gadient Descent
- In BGD we split the dataset into mini batches, e.g our dataset has 100 data points we split it into 5 batches, the weights and biases are updated after each mini batch is processed. 

- For example, our data set has 100 datapoints, we split it into 10 batches, so 10 times weights and biases will be update per epoch. Our epochs are 5, so our weights and biases will be updated 50 times.
