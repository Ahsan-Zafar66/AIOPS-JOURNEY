# vanishing Gradient
- In simple words vanishing gradient means when there is extremely small change in partial derivative of loss with respect to weight and due to which there is no significant change in the loss function and neural network fails to train more, as we cannot reduce loss further

# How to recognize vanishing gradient
- When there is no change in loss over many epochs
- Plot a graph of weight, its values vs epoch number and observe if the grapgh is consistent, which will tell that there is no change in values over the epochs.


# How to handel vanishing gradient?
- Reduce Model Complexity, but neural network will not be able to learn the complex hidden patterns. It can become handy but not used mostly.
- Use different activation function. You can use ReLu. For dying relu probelm use leaky relu
- Proper weight initialization. for example Gloret, xavier.
- Batch Normalization, it is a layer
- Residual network.
