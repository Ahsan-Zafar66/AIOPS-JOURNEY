# Perceptron
- Perceptron is that it randomly selects a data point and see if it is in its right region mean if positiv than in positive region and if negative class than in negative region. If the positive class data point is in negative region it moves the line to correctly classifiy the data point. So, we repeat this process 1000 times and each time a random point is take and line is moved if the data point is in the wrong region. This trick works but it have some problems

# Problems with perceptron trick
- The line we got are we sure that it is the best line?
- The number of  times we iterate 1000 times what if it selects the always the data point that is in correct region and line does not moves or improve.

# Weights
- Weights tell you feature importance, how much a feature is important in the prediction of a target variable

# Perceptron Problems
- it cannot work on non-linear data, it can only work on linear separable data
