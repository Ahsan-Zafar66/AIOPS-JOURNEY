```python
import numpy as np
from keras.layers import Embedding
from keras.preprocessing.text import Tokenizer
import tensorflow
from tensorflow import keras
from keras.utils import pad_sequences
```

```python
data = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
# oov means out of vocabulary, during prediction of new data if a word comes which is not present in training data replace it with <nothing>
# it is done to deal with words not present in training data and to make predicitons on new data smoothly
tokenizer = Tokenizer(oov_token="<nothing>")
```

```python
# let's apply tokenizer on our data
tokenizer.fit_on_texts(data)
```

```python
# getting the integer encoding
tokenizer.word_index
```

- we can see each word has been assigned a numerical number.

```python
# let's count the occurence of each word
tokenizer.word_counts
```

```python
# let;s check how many data points or sentences in my data
tokenizer.document_count
```

- so, we have 10 sentences in our data.

```python
# now, lets convert the text into numerical data according to the number they got
sequences = tokenizer.texts_to_sequences(data)
sequences
```

- As we can see, each sentence or data point is of different length, so we will aplly padding on it.

```python
# let's do padding
# post means add zero after
# we can also use pre, which will add zero before
padded_sequences = pad_sequences(sequences,padding="post")
```

```python
padded_sequences
```

- now we can see each sentence or data point is of equal length, we have done zero padding here.

# APPLYING SIMPLERNN ON IMBD DATA
- Remeber we are using integer encoding technique

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Flatten
```

```python
# loading the imbd data
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
# lets study the shapes of data
X_train.shape
```

```python
y_train.shape
```

```python
# let's check targe
y_train[:10]
```

- 1 means positive review and 0 means negative review.

```python
# now let's explore features
X_train[0]
```

```python
# let's find if the data has different lengths
print("Length of first review in training data {}".format(len(X_train[0])))
print("Length of second review in training data {}".format(len(X_train[1])))
```

- As we can see lengths are different so we will apply zero padding. This data is already preprocessed means that text is converted into integer encoding

```python
# padding the training and testing data
# there is a parameter called maxlen which will return a define dlength of review, like if I set maxlen = 100, all of my reviews will be of 100 words
train_pad = pad_sequences(X_train,padding="post",maxlen=50)
test_pad = pad_sequences(X_test,padding="post",maxlen=50)
```

```python
# let's find if the data has same lengths
print("Length of first review after padding in training data {}".format(len(train_pad[0])))
print("Length of second review after padding in training data {}".format(len(train_pad[1])))
```

- Now, all of our reviews has same length


# Building RNN model

```python
train_pad[0].shape
```

```python
# # (2494,1) is 2494 timesteps and 1 is input features, so for each review one word will enter the network so total timesteps for one review are 2494 and feature is 1 as only one word is inputted
model = Sequential()
model.add(SimpleRNN(32,input_shape=(50,1),return_sequences=False))
model.add(Dense(1,activation="sigmoid"))
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
model.fit(train_pad,y_train,epochs=5,validation_split=0.2)
```

```python
from keras.datasets import imdb
from keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
model = Sequential()
model.add(Embedding(10000,output_dim= 2,input_length = 50))
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_test.shape
```

```python
y_test.shape
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```
