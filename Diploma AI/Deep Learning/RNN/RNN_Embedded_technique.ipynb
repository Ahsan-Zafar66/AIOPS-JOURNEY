```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense, Embedding, SimpleRNN
from keras.utils import pad_sequences
from keras.datasets import imdb
from keras import Sequential
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
model = Sequential()
# 10000 phlay parameter ka mtlb hai ky 10000 words ki vocabulary use karo beshak data mn zyada hain
## 10000 is that I want to keep 10,000 words vocabulary whatever the real number is I will keep only 10,000
# output_dim =2 means ky phla jo data point hai, uska phla word 10,000 dimensional vector mn convert ho ga or embedding layer mn enter ho ga, output 2 hai ky hr word jo enter ho ga usy 2 vectors define karein gein (it is a hyperparameter)
# input length means the length of data point entering the embedding layer
model.add(Embedding(10000,output_dim= 2,input_length = 50))
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```
