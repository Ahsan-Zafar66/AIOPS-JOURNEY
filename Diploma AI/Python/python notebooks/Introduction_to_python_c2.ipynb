# List Methods

```python
# append
# Description: Adds elements at the end of a string
numbers = [1, 2, 3]
numbers.append("orange")
print(numbers)  # Output: [1, 2, 3, 4]
```

```python
# extend()
# Description: Appends elements from another iterable (list, tuple, etc.) to the end of the list.
fruits = ["apple", "banana"]
more_fruits = ["orange", "grape"]
fruits.extend(more_fruits)
print(fruits)  # Output: ['apple', 'banana', 'orange', 'grape']
```

```python
# insert()
# syntax insert(position to insert data, data)
# Description: Inserts an element at a specified index in the list.
numbers = [1, 2, 4, 5]
numbers.insert(2, 3)
print(numbers)  # Output: [1, 2, 3, 4, 5]
```

```python
# remove()
# remove(data we want to remove)
# Description: Removes the first occurrence of a specified value from the list.
numbers = [1, 2, 3, 4, 3]
numbers.remove(3)
print(numbers)  # Output: [1, 2, 4, 3]
```

```python
# pop()
# Description: Removes and returns the element at a given index. If no index is specified, it removes and returns the last element.
numbers = [1, 2, 3, 4, 5]
popped_element = numbers.pop()
print(popped_element)  # Output: 3
print(numbers)  # Output: [1, 2, 4, 5]
```

```python
# index()
# Description: Returns the index of the first occurrence of a specified value in the list.
numbers = [10, 20, 30, 20, 40, 20]
index = numbers.index(20)
print(index)  # Output: 1
```

```python
# count()
# Description: Returns the number of occurrences of a specified element in the list.
numbers = [1, 2, 2, 3, 2, 4]
count = numbers.count(2)
print(count)  # Output: 3
```

```python
# sort()
# Description: Sorts the elements of the list in ascending order. (Note: For strings, it sorts alphabetically.)
numbers = [4, 2, 1, 3]
numbers.sort()
print(numbers)  # Output: [1, 2, 3, 4]
```

```python
# reverse()
# Description: Reverses the order of elements in the list.
numbers = [1, 2, 3, 4]
numbers.reverse()
print(numbers)  # Output: [4, 3, 2, 1]
```

```python
# clear()
# Description: Removes all elements from the list, making it empty.
numbers = [1, 2, 3, 4]
numbers.clear()
print(numbers)  # Output: []
```

```python
# copy()
# Description: Creates a shallow copy of the list.
original_list = [1, 2, 3]
copied_list = original_list.copy()
print(copied_list)  # Output: [1, 2, 3]
```

# Dictionaries
- In Python, a dictionary is a powerful data structure used to store a collection of key-value pairs. Unlike sequences like lists that are indexed by a range of numbers, dictionaries are indexed by keys, which can be any immutable type (strings, numbers, tuples). Keys within a dictionary are unique and each key maps to a specific value.

# Importance of Dictionaries in Python:
- **Fast Retrieval:** Dictionaries use hash tables, allowing for quick retrieval of values based on keys, even for large collections.

- **Flexibility:** They can store different types of data (integers, strings, lists, other dictionaries) as values.

- **Key-Value Mapping:** Dictionaries enable efficient mapping between keys and corresponding values, making it easy to retrieve information based on a specific identifier.

# Methods

```python
# get()
# Description: Retrieves the value for a given key, returns None if the key is not found.
employee_details = {'001': 'Alice', '002': 'Bob', '003': 'Charlie'}
name = employee_details.get('002')
print(name)  # Output: Bob
```

```python
# accessing info directly
employee_details["001"]
```

```python
# accessing information through get method
employee_details.get("001")
```

```python
# adding data in a dictionary
employee_details["004"] = "Jhon"
```

```python
# keys()
# Description: Returns a view object containing the dictionary's keys.
employee_details = {'001': 'Alice', '002': 'Bob', '003': 'Charlie'}
keys = employee_details.keys()
print(keys)  # Output: dict_keys(['001', '002', '003'])
```

```python
# values()
# Description: Returns a view object containing the dictionary's values.
employee_details = {'001': 'Alice', '002': 'Bob', '003': 'Charlie'}
values = employee_details.values()
print(values)  # Output: dict_values(['Alice', 'Bob', 'Charlie'])
```

```python
# items()
# Description: Returns a view object containing the dictionary's key-value pairs as tuples.
employee_details = {'001': 'Alice', '002': 'Bob', '003': 'Charlie'}
items = employee_details.items()
print(items)  # Output: dict_items([('001', 'Alice'), ('002', 'Bob'), ('003', 'Charlie')])
```

```python
# pop()
# Description: Removes and returns the value for a given key. If the key is not found, it raises a KeyError (or returns a specified default value).
employee_details = {'001': 'Alice', '002': 'Bob', '003': 'Charlie'}
removed_value = employee_details.pop('002')
print(removed_value)  # Output: Bob
```

```python
# popitem()
# Description: Removes and returns the last inserted key-value pair as a tuple.
employee_details = {'001': 'Alice', '002': 'Bob', '003': 'Charlie'}
removed_item = employee_details.popitem()
print(removed_item)  # Output: ('003', 'Charlie')
```

```python
# update()
# Description: Updates the dictionary with key-value pairs from another dictionary or iterable.
employee_details = {'001': 'Alice', '002': 'Bob'}
new_employees = {'003': 'Charlie', '004': 'David'}
employee_details.update(new_employees)
print(employee_details)  # Output: {'001': 'Alice', '002': 'Bob', '003': 'Charlie', '004': 'David'}
```

```python
# clear()
# Description: Removes all items from the dictionary.
employee_details = {'001': 'Alice', '002': 'Bob', '003': 'Charlie'}
employee_details.clear()
print(employee_details)  # Output: {}
```

```python
# copy()
# Description: Creates a shallow copy of the dictionary.
employee_details = {'001': 'Alice', '002': 'Bob'}
copied_dict = employee_details.copy()
print(copied_dict)  # Output: {'001': 'Alice', '002': 'Bob'}

```

```python
# fromkeys()

# Description: Creates a new dictionary with keys from a sequence and values set to a default value.

# yo can use this method when you want to give same value to you keys
keys = ['a', 'b', 'c']
default_value = 0
new_dict = dict.fromkeys(keys, default_value)
print(new_dict)  # Output: {'a': 0, 'b': 0, 'c': 0}
```

```python
# creating a list of employees
employees = ["Jhon","Smith","Caroline"]

# present value
present  = "P"

# other employees list
employees_other = ["Klaus","Paul","Steve"]

# leave value
leave = "L"

# present employees information
present_employee_dict = dict.fromkeys(employees,present)

# employees on leave
leave_employee_dict = dict.fromkeys(employees_other,leave)
```

# if-else Conditional Statements
- In Python, like in many programming languages, if-else conditions are fundamental for controlling the flow of a program based on certain conditions. These conditions allow you to execute specific blocks of code when certain criteria are met or to handle different scenarios based on the evaluation of expressions.

# Importance of if-else Conditions:
- **Conditional Execution:** if-else conditions enable the execution of specific blocks of code based on whether a condition evaluates to True or False. This allows programs to perform different actions depending on input, user interactions, or certain states.

- **Control Flow:** They allow for branching in the program flow, facilitating decision-making processes. For example, you can handle different cases in your program based on user input or system states.

- **Error Handling:** if-else conditions are essential for error handling and validation. They help ensure that certain conditions are met before executing critical parts of the code, preventing errors or unexpected behaviors.



```python
if condition:
    # Code block executed if condition is True
    statement1
    statement2
    # ...

else:
    # Code block executed if condition is False
    statement3
    statement4
    # ...
```

```python
# example 1 checks if the number is positive or negative
num = 10

if num >= 0:
    print("The number is positive.")
else:
    print("The number is negative.")
```

```python
# example # 2
# Determine if a Number is Even or Odd:

num = 7

if num % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

```python
# example 3
# Compare Two Numbers:
a = 5
b = 10

if a > b:
    print("a is greater than b.")
else:
    print("b is greater than a.")
```

```python
# Check if a String is Empty:
string = "Hello"

if string:
    print("The string is not empty.")
else:
    print("The string is empty.")
```

```python
# Compare String Values:

name = "Alice"

if name == "Alice":
    print("Welcome, Alice!")
else:
    print("You are not Alice.")
```

```python
# Verify User's Age:

age = 20

if age < 18:
    print("You are underage.")
else:
    print("You are an adult.")
```

```python
# Check Membership in a List:
fruits = ['apple', 'banana', 'orange']

if 'banana' in fruits:
    print("Banana is in the list.")
else:
    print("Banana is not in the list.")
```
