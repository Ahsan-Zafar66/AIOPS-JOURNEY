# List Merging

```python
def merge_lists(*lists):
    final_list = []
    for data in lists:
      #print(data)
      final_list += data
    return final_list
```

```python
merge_lists([1,2],[4,5,6],[7,8,9,10])
```

```python
# a function that takes in two numbers and performs addition, subtraction, multiplication and division on them
def perform_operations(a, b):
    addition = a + b
    subtraction = a - b
    multiplication = a * b
    if b != 0:
        division = a / b
    else:
        division = "Cannot divide by zero"

    return addition, subtraction, multiplication, division


```

```python
# multiple outputs of a function are stored as tuple if the variable saving them is one
results =  perform_operations(a=10,b=10)
```

```python
results
```

```python
# multiple outputs of a function can be stored as multi variables
sum, difference, multiply, divide = perform_operations(a=5,b=0)
```

```python
# to make a password generator of varying length
def random_password_generator(length):
  string = "abcdefghijklmnopqrstuvwxyz!@#$%^&*()"
  numbers = 12345678910
  space = string + str(numbers)
  password_length = length
  my_password = ""
  for i in range(0,password_length):
    random_letter = random.choice(space)
    my_password += random_letter

  return my_password
```

# Guessing Game

```python
import random

def guess_the_number():
    number = random.randint(1, 100)
    attempts = 0
    while True:
        guess = int(input("Enter your guess (between 1 and 100): "))
        attempts += 1
        if guess < number:
            print("Too low, try again!")
        elif guess > number:
            print("Too high, try again!")
        else:
            print(f"Congratulations! You guessed the number {number} in {attempts} attempts.")
            break

# Play the game
guess_the_number()
```
