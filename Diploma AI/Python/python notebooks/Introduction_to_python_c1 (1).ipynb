# Python
- Python is a high-level, interpreted, and general-purpose programming language that emphasizes readability and ease of use. It was created by Guido van Rossum and first released in 1991. Python has become one of the most popular programming languages due to its simplicity, versatility, and a large and active community.

- Here are some key characteristics and features of Python:

- **Readable and Expressive Syntax:** Python code is designed to be readable and expressive, making it easier for developers to write and   maintain code. The language enforces the use of indentation to define code blocks, which enhances readability.
- **Interpreted Language:** Python is an interpreted language, which means that code can be executed directly without the need for compilation. This contributes to a quicker development cycle, as there is no separate compilation step.
- **Dynamic Typing:** Python is dynamically typed, meaning that variable types are determined at runtime. This allows for more flexibility in coding but requires careful attention to variable types during development.

- **Object-Oriented Programming (OOP):** Python supports object-oriented programming principles, allowing developers to create and use classes and objects. Encapsulation, inheritance, and polymorphism are key concepts in Python's object-oriented paradigm.

- **Standard Library:** Python comes with a comprehensive standard library that includes modules and packages for a wide range of tasks, from working with data structures to handling network protocols, file I/O, and more. This reduces the need for developers to write code from scratch for common functionalities.
- **Extensive Third-Party Libraries:** Python has a rich ecosystem of third-party libraries and frameworks that cater to various domains such as web development (Django, Flask), scientific computing (NumPy, SciPy), machine learning (TensorFlow, PyTorch), data analysis (pandas), and more.

- **Cross-Platform Compatibility:** Python is a cross-platform language, meaning that Python code can run on various operating systems with little to no modification. This makes it highly portable and adaptable.

- **Community and Documentation:** Python has a vibrant and supportive community of developers. The community contributes to the language's growth, development, and the creation of a wealth of documentation and tutorials, making it accessible to developers of all skill levels.
- **Versatility:** Python is a versatile language suitable for a wide range of applications, including web development, data science, artificial intelligence, automation, scripting, and more. Its versatility contributes to its popularity across different domains.
- **Open Source:** Python is an open-source language, and its source code is freely available. This encourages collaboration, transparency, and a sense of community ownership.

# Integer or int
- In Python, the int data type is used to represent integers, which are whole numbers without any fractional part. Python allows integers to be positive, negative, or zero. The int type in Python has unlimited precision, meaning that integers can be arbitrarily large (limited only by available memory).

- In python int data type is mostly used for calculation purpose.

# Arithmetic Operations:
- Integers support standard arithmetic operations like addition, subtraction, multiplication, and division.

```python
# example of integers
# Creating integers
a = 5
b = -10
c = 0

# Printing integers
print(a, b, c)
```

```python
# calculation in integers

# additon
addition_ab = a + b

print(addition_ab)
```

```python
# subtraction
subtraction_ab = a - b

print(subtraction_ab)
```

```python
# multiplication
multiplication_ab = a * b

print(multiplication_ab)
```

```python
# division
division_ab = a / b

print(division_ab)
```

```python
# manual way
power_a = a ** 3
power_a
```

```python
# power using method
power_ac = pow(a, 3)
print(power_ac)
```

# Float
- In Python, the float data type is used to represent floating-point numbers, which are numbers that have both an integer and a fractional part. Floats are used to represent real numbers and can be positive, negative, or zero.

```python
# Creating floats
x = 3.14
y = -0.5
z = 2.0

# Printing floats
print(x, y, z)
```

```python
value = 4.55555588889999944400994
print(round(value,))
```

```python
# Round

# The round() function rounds a float to the nearest integer or a specified number of decimal places.

rounded_x = round(x)
rounded_x_2_decimals = round(x, 2)

print(rounded_x)
print(rounded_x_2_decimals)
```

```python
# Absolute Value

# The abs() function returns the absolute value of a float.
absolute_value_y = abs(y)
print(absolute_value_y)
```

# Comparison Operators
Comparison operators in Python are used to compare two values and produce a boolean result indicating the relationship between them. These operators are commonly used in conditional statements, loops, and other situations where you need to make decisions based on the comparison of values

# Types:
- Equal to (==)
- Not Equal to (!=)
- Greater than (>)
- Less than (<)
- Greater than or equal to(>=)
- Less than or equal to (<=)

```python
# Equal To example

a = 5
b = 7

result = (a == b)
print(result)
```

```python
# Not Equal to example
a = 5
b = 7

result = (a != b)
print(result)
```

```python
# greater than example
a = 5
b = 7

result = (a > b)
print(result)
```

```python
# less than
a = 5
b = 7

result = (a < b)
print(result)
```

```python
# greater than or equal to
a = 5
b = 5

result = (a >= b)
print(result)
```

```python
# less than or equal to
a = 5
b = 7

result = (a <= b)
print(result)
```

# Tuple
- A tuple is an ordered and immutable collection of elements in Python. It is similar to a list, but once a tuple is created, you cannot modify its contents, you can't add, remove, or modify elements. Tuples are defined using parentheses ().

```python
# Creating a tuple
my_tuple = (1, 2, 3, 'a', 'b', 'c',10.2,True)

# Accessing elements
print(my_tuple[-3])
print(my_tuple[3])
```

```python
# count() Method:

# Returns the number of occurrences of a specified element in the tuple.
my_tuple = (1, 2, 3, 1, 2, 3, 1, 2, 3)
count_2 = my_tuple.count(2)
print(count_2)
```

```python
# count example of tuple with string data in it
my_tuple_string = ("ahsan","smith","python","ahsan")
my_tuple_string.count("book")
```

```python
# index() Method:

# Returns the index of the first occurrence of a specified element in the tuple.
my_tuple = (10, 20, 30, 40, 50)
index_30 = my_tuple.index(40)
print(index_30)
```

```python
# Concatenation (+):

# Combines two or more tuples.
tuple1 = (1, 2, 3)
tuple2 = ('a', 'b', 'c')
combined_tuple = tuple1 + tuple2
print(combined_tuple)
```

```python
# Repetition (*):

# Repeats a tuple a specified number of times.
my_tuple = (1, 2, 3)
repeated_tuple = my_tuple * 3
print(repeated_tuple)
```

```python
# in example of tuple with string data
my_tuple8 = ("ahsan","colab","book")
"ahsan" in my_tuple8
```

```python
# Membership Test (in and not in):

# Checks if an element is present in the tuple.
my_tuple = (10, 20, 30, 40, 50)
check_30 = 30 in my_tuple
print(check_30)
```

```python
# Maximum and Minimum (max() and min()):

# Returns the maximum and minimum values in a numeric tuple.
numeric_tuple = (10, 20, 5, 30, 15)
max_value = max(numeric_tuple)
min_value = min(numeric_tuple)
print("Maximum value is:",max_value)
print("Minimum value is:", min_value)
```
