# Loops
- Loops are fundamental constructs in programming, allowing you to execute a block of code repeatedly. Here's an explanation of the for and while loops in Python along with ten basic examples for each:

# for Loop:
- The for loop iterates over a sequence (such as a list, tuple, string, or range) and executes the block of code for each element in the sequence until the sequence is exhausted.

```python
# experimenting for loop with string and counting how many times it iterates
my_string = "Ahsan Zafar"

count = 0
for i in my_string:
  print(i)
  count +=1

print("Number of times for loop ran",count)
```

```python
# for loop with lists
fruits = [10,True, 'banana', 'orange',"grapes","watermelon","pineappled",20]
for fruit in fruits:
  if type(fruit) == str:
    print(fruit.upper())
    print("-" * 50)
  else:
    print("The datatype is not string",)
    print("Data is ",fruit)
    print("Type of data is",type(fruit))
    print("-" * 50)


```

```python
# iterating using range
for num in range(5):
    print(num)
```

```python
# for loop on strings
word = "Python"
for letter in word:
  print("Ahsan")
  print(word)
  print(letter)
```

```python
# example of for loop with tuple
# when there is only one  tuple in nested tuples, it runs according to the data inside that single tuple
# when there is more than one tuple inside multiple tuples, the loop will iterate as many times as there are number of tuples

# example of nested tuples with only one tuple inside
coordinates = ((3,5,7))
count = 0
for coordinate in coordinates:
    print(coordinate)
```

```python
# example of for loop with tuple
# when there is only one  tuple in nested tuples, it runs according to the data inside that single tuple
# when there is more than one tuple inside multiple tuples, the loop will iterate as many times as there are number of tuples

# example of nested tuples with multiple tuples inside
coordinates = ((3,5,7),(9,11,13))
count = 0
for coordinate in coordinates:
    print(coordinate)
```

```python
# for loop with dictionaries
student = {'name': 'Alice', 'age': 20, 'grade': 'A'}
for key,value in student.items():
  print(key)
  print(value)
  print("-" * 50)




```

```python
# for loop example with enumerate
# loop  names that are available inside loop contains the data of last data point on which it ends
words = [('apple',10,"good"), ('banana',11,"rotten"), ('orange',12,"good")]
for fruit,num,quality in words:
  print("Fruit is ",fruit,"and there are ",num,"quality of fruit is",quality)


```

```python
# nested for loop with iteration
for i in words:
  for x, y, z in [i]:
    print(x,y,z)
    print("-" * 50)
```

# while Loop:
- The while loop executes a block of code repeatedly as long as a specified condition is True. It continues until the condition becomes False.

```python
# while loop exampl breaking the code whn the temperatre is 100
temperature = 50
while temperature < 100:
  print("Boiler is running and temperature is {}".format(temperature))
  temperature += 1
```

```python
# example of while loop with count
count = 0
while count < 5:
    print(count)
    count += 1

```

```python
# example of while loop with number
number = 1
while number <= 10:
    print(number)
    number += 2

```

```python
# while loop example with if condition
x = 5
while x > 0:
    print(x)
    x -= 1
    if x == 2:
        break

```

```python
count = 0
while count < 3:
    print("Inside while loop")
    count += 1
else:
    print("Outside while loop")

```

```python
# Running while loop until a defined count
count = 0
while True:
    print("This is an infinite loop")
    count += 1

    if count == 1000:
      print("While loop ran 1000 times")
      break


```

```python
# password checker using while loop
password = 'abc123'
while True:
    user_input = input("Enter password: ")
    if user_input == password:
        print("Access granted!")
        break
    else:
        print("Wrong password. Try again.")

```

```python
# continue statement skips the code after it and move to the next iteration
# that is why 3 is not printed
x = 0
while x < 5:
    x += 1
    if x == 4:
        continue
    print(x)


```

```python
x = 0
while x < 3:
    y = 0
    while y < 2:
        print(x, y)
        y += 1
    x += 1

```

```python
num = 5
factorial = 1
while num > 0:
    factorial *= num
    num -= 1
print("Factorial {}".format(factorial))

```
