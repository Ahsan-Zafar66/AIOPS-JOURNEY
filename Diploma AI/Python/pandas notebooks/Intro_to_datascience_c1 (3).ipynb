# What is Data Science?
- Data science is an interdisciplinary academic field that uses statistics, scientific computing, scientific methods, processes, algorithms and systems to extract or extrapolate knowledge and insights from potentially noisy, structured, or unstructured data. (wikipedia)
- Data science combines math and statistics, specialized programming, advanced analytics, artificial intelligence (AI), and machine learning with specific subject matter expertise to uncover actionable insights hidden in an organization's data. These insights can be used to guide decision making and strategic planning. (IBM)

# Python Libraries For Data Science
- Some of the most famous and widely used Python libraries in data science include:

- **NumPy:** Fundamental library for numerical computing in Python. It provides support for arrays, matrices, mathematical functions, and linear algebra operations.

- **Pandas:** A versatile library for data manipulation and analysis, offering powerful data structures like DataFrame and Series, and tools for data cleaning, transformation, and exploration.

- **Matplotlib:** A comprehensive plotting library for creating static, interactive, and publication-quality visualizations, ideal for data visualization and exploration.

- **Seaborn:** Built on top of Matplotlib, Seaborn offers a high-level interface for creating attractive statistical graphics. It simplifies the creation of complex visualizations and supports exploratory data analysis.

- **Scikit-learn:** An extensive machine learning library providing simple and efficient tools for data mining and analysis. It offers a wide range of algorithms for classification, regression, clustering, dimensionality reduction, and more.

- **SciPy:** Built on top of NumPy, SciPy is a library for scientific and technical computing. It includes modules for optimization, integration, linear algebra, signal processing, and other scientific computing tasks.

- **TensorFlow and PyTorch:** Deep learning frameworks used for building and training neural networks. TensorFlow and PyTorch are widely adopted for tasks such as image recognition, natural language processing, and more.

- **Statsmodels:** A library that supports the estimation of statistical models, hypothesis testing, and statistical data exploration.

- **NLTK (Natural Language Toolkit):** A powerful library for working with human language data. It provides tools for text analysis, tokenization, stemming, tagging, parsing, and more.

- **Keras:** An open-source neural network library built on top of TensorFlow. Keras offers a user-friendly interface for designing, training, and deploying deep learning models.

# Data Structures in Pandas
- **Series:**
- A Series in Pandas is a one-dimensional labeled array that can hold various data types. It consists of two main components: the data and the index. The data can be of any data type, including integers, floats, strings, or even Python objects. The index is a set of labels that uniquely identifies each element in the Series. By default, the index starts from 0 but can be customized to use specific labels.



```python
# never mix datatypes in a single column

# creating a series in data
data = [1,2,3,4,10]
series = pd.Series(data,name="integers")

```

```python
# Creating a Series with default numeric index
data = [10, 20, 30, 40, 50]  # Sample data
series_default_index = pd.Series(data,name="data")
#print("Series with default numeric index:")
series_default_index
```

```python
# creating a series with custom index to fetch data
data = [10,20,30,40,50,60]
indexes = ["ten","twenty","thirty","fourty","fifty","sixty"]

# name = name you want to give to your series
# index = you have to pass a list equaling your data points of indexe
index_series = pd.Series(data,name="numbers",index=indexes)
index_series["twenty"]
```

```python
# Creating a Series with custom labels as index
data = [10, 20, 30, 40, 50]  # Sample data
index_labels = ['A', 'B', 'C', 'D', 'E']  # Custom index labels
series_custom_index = pd.Series(data, index=index_labels)
print("Series with custom index labels:")
print(series_custom_index)
```

# Pivot Tables

```python
# code to create dummy data of dogs

import random

# Provided data for breeds and colors
breeds = ['Labrador', 'Poodle', 'Chow Chow', 'Schnauzer', 'Chihuahua', 'St. Bernard']
colors = ['Brown', 'Black', 'Gray', 'Tan', 'White']
names = ["Max","Stella","Charlie","Stafy","gracie"]

# Generating data for 30 data points
data = {
    'breed': [random.choice(breeds) for _ in range(40)],
    'color': [random.choice(colors) for _ in range(40)],
    'height_cm': [random.randint(20, 80) for _ in range(40)],  # Random height between 20 and 80
    'weight_kg': [random.uniform(1, 40) for _ in range(40)],    # Random weight between 1 and 40
    'name': [random.choice(names) for _ in range(40)]
}

# Creating a DataFrame
dogs_new = pd.DataFrame(data)
```

```python
# creating pivot tables
# index, the column I want to keep as index
# columns, the column whose values I want to keep as columns
# values, the column which values will fill the table
pivot_table = dogs_new.pivot_table(index="breed",columns="color",values="height_cm")
```

```python
dogs_new.head()
```

```python
dogs_new["color"].unique()
```

```python
pivot_table
```

```python
# creating pivot table again but replacing NaN with 0
pivot_table2 = dogs_new.pivot_table(index="breed",columns="color",values="height_cm",fill_value=0)
```

```python
pivot_table2
```

```python
# average height of each breed
pivot_table2.mean(axis=1)
```

```python
# average height of each color
pivot_table2.mean(axis=0)
```

```python
# pivot table having averages acrros margins
pivot_table3 = dogs_new.pivot_table(index="breed",columns="color",values="height_cm",fill_value=0,margins=True,aggfunc="mean")
```

```python
pivot_table3
```

```python
pivot_table_max_height = dogs_new.pivot_table(index="breed",columns="color",values="height_cm",fill_value=0,margins=True,aggfunc="max")
```
