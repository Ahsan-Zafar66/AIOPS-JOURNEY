```python
# defining a string
name = "AHSAN zafar"
```

```python
# to get all the methods of a string
print(dir(name))
```

```python
# casefold advance form of lower
name.casefold()
```

```python
# getting help about  a particular function
help(str.casefold)
```

```python
# center: center a string
print(name.center(20))
```

```python
# checking if the string ends with a desired letter
name.endswith("r")
```

```python
# title: capitalizes first letter of each word
name.title()
```

```python
# zero filling the name
name.zfill(20)
```

```python
# it swaps the order, turns capital into lower and lower into capital
name.swapcase()
```

```python
# split: splits the string at the mentioned letter
name.strip(" ")
```

```python
# strip does not remove the middle word, it only removes starting and ending
"Ahsan Zafar Warraich".strip("Zafar")


# strip has variants rstrip strip from right side
# left strip strips from left side
```

```python
# index : tells the index of letter
name.index("A")
```

```python
# join: "whatto join".join(iterable)
# join the letters in the string with a space in each of them
" ".join(name)
```

```python
# split by default conver a string into list and splits on space
name.split()
```

```python
# if you split by something it will be removed and a single quotation is stored on its place
name.split("AHSAN")
```

```python
# replace
mystring = "Ahsan$ Zafar$ Warraich$"
mystring.replace("$","")
```
