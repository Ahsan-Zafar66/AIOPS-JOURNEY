# Classes
- Classes have two things attributes and behaviors
- Attributes means variables which are defined in __init__ function by using self. For example self.name.
- Behaviours are functions which are methods of class.

```python
# classes syntax
class class_name:

  def __init__(self,variables):
    self.variable = variable_name

  def method_name(self):
    print(self.variable)
```

```python
# defining a class that has two functions in it
# first function show the computer processes
# second show if it can be used for DL or not

# DEFINING CLASS
class check:

  # defining variables
  def __init__(self,ram,cpu):
    self.ram  = ram
    self.cpu = cpu

  # defining a method
  def computer(self):
    print("CPU is {} and Ram is {}".format(self.cpu,self.ram))

  # defining another method
  def compatible(self):
    if self.ram > 8 and self.cpu > 2:
      print("It can be used for DL")
    else:
      print("It cannot be used for DL")
```

```python
# defining the specs
# using an object for class
myspecs = check(cpu=3,ram=16)

```

```python
# calling the object (as class is stored in object with values) and methods
print(myspecs.computer())
#print(myspecs.compatible())
```

# Above Class in a different way
- In this method or way, you would be passing argument each time you call a method but in previous method you just pass arguments to class one time and can call methods they will directly fetch arguments from the class defined in it.

```python
# the above class can also be defined differently rather than passing arguments through an object we can call method directly by passing arguments

class check:

  # defining variables
  def __init__(self,ram,cpu):
    self.ram  = ram
    self.cpu = cpu

  # defining a method rather than passing self like above we can pass directly arguments
  def computer(ram,cpu):
    print("CPU is {} and Ram is {}".format(cpu,ram))

  # defining another method
  def compatible(ram,cpu):
    if ram > 8 and cpu > 2:
      print("It can be used for DL")
    else:
      print("It cannot be used for DL")
```

```python
# now calling methods directly
check.computer(ram=9,cpu=3)
```

```python
# now calling compatible method directly
check.compatible(ram=9,cpu=3)
```

# Accessing Variables and Changing their value

```python
# defining class in an object
base = check(ram=9,cpu=3)
# printing value
print(base.ram)


# let's change the value and print it
base.ram=20
print(base.ram)
```

```python
# defining a class
class DataBase:

  # defining variables of the class
  def __init__(self,name,age):
      self.name = name
      self.age = age

  # defiing a method
  def data(self):
    print("{} is learning Classes right now and he is {} years old".format(self.name,self.age))

```

```python
# giving values in class
base = DataBase("Ahsan",24)

# calling method data through object base
base.data()
```

```python
# giving values in class
base = DataBase(name="Ahmed",age=26)

# calling method data through object base
base.data()
```
