```python
# list comprehension
my_list = [(name,age) for name,age in zip(["ahsan","ahmed"],[24,26])]
my_list
```

```python
my_list = [(name,age) for name,age in zip(["ahsan","ahmed"],[24,26]) if name == "ahsan"]
my_list
```

```python
# dictionary comprehension
my_dict = {key:value for key,value in zip(["ahsan","ahmed"],[24,26])}
```
