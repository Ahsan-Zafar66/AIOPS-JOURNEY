```python
# inserting line at the top of file or at desired line
# first we open the file, read it's lines into a list, then enter data at specific line or index in the list
# and write the file containing new line at desired position
# inserting a line at the top of file

# Opening the file and read its contents into a list
with open("/content/Lecture # 42 YUM PATCH.txt","r") as file:
    lines = file.readlines()

# Insert the new line at the desired position (e.g., line number 3)
new_line = "This is the new line at line 0.\n"
lines.insert(0, new_line)  # Remember, list index starts at 0

# Rewriting the file with the new contents
with open("/content/Lecture # 42 YUM PATCH.txt", 'w') as file:
    file.writelines(lines)
```
