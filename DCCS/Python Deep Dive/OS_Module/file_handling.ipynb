#### Adding new data at desired line a file

```python
# inserting a data at desried line
# step # 1, open the file and perform readlines function, it will save the file lines in a list
# now you can insert your desired text at the desired index in this list
with open("mytextfile.txt","r") as file:
    lines = file.readlines()
mytext = "my name is ahsan and this is my text file"
lines.insert(0,mytext)


```

```python
# now open the file again and write the list containing new and old lines
with open("mytextfile.txt","w") as file:
    file.writelines(lines)
```

```python
# now open the file again and read lines
with open("mytextfile.txt","r") as file3:
    for line in file3.readlines():
        print(line)
```

#### Reading JSON Files
- JSON stands for Jave Script Object Notation
- It is  data format which is structure data

```python
import json
```

```python
# Step # 1: read the json file as normal file
# step # 2 use the json_load function and pass the read file in it
with open("first_json.json","r") as js:
    content = json.load(js)

print(type(content))
```

```python
# unwrapping the nested dictionaries inside the list 
for i in content:
    for k,v in i.items():
        print(k,v)
```

```python
# creating a json file and dumping data in it
import json

myinfo = {
	"server1" : {
		"IBM": {
			"datacenter":"Bangalore",
			"env": {
				"PR": "192.168.1.1",
				"DR": "192.168.1.2"
				}
			}
		    }
	}
print(type(myinfo))

mypath="user_details.json"
with open(mypath,'w') as df:
    # indent is for spacing
    json.dump(myinfo,df,indent=4)
```

```python
# loading fruits.json and listing fruits having prive greater than 1
with open("fruits.json","r") as jf:
    fruits = json.load(jf)
```

```python
count =0 
fruit_qualify = []
for i in fruits:
    if i["price"] >=1:
        #print("Fruit Name:",i["name"])
        count+=1
        fruit_qualify.append(i["name"])
print(f"Total Number of fruits having price greater than $1 are {len(fruit_qualify)}")
print("Fruits are",", ".join(fruit_qualify))
```

#### Reading csv file using csv module

```python
import csv

mypath=r"file_path_of_csv_file"

# open the csv file
with open(mypath) as datafile:
    
    # as in json we use load for csv files we use reader, reader gives us an iterator object, we can use for loop to see data of csv file
    # delimiter can be different in different files so you can specify using the delimiter keyword argument
    data=csv.reader(datafile,delimiter=",")
    
    # running a for loop to see the data
    for each in data:
        print(each)
```

```python
# writing data in a csv file
import csv
line1 = ["Name","Age","Gender"]
line2 = ["Ahsan",24,"Male"]

# python 3 by defaulte create an empty line after writing each line, so we can set newline="", not no new line will be created after writing each line

# w is for wrting (overwriting on previous content), "a" is for appending below the previous content
# for python 2 remove newline and instead of just "w" use "wb" now no new line will be created
with open("mycsv.csv","w",newline="") as csv_file:
    # now file has become write object we can wrtie row by row data in it
    file = csv.writer(csv_file)
    file.writerow(line1)
    file.writerow(line2)
```

```python
# reading the above created file
with open("mycsv.csv") as read_csv:
    read_object = csv.reader(read_csv)
    for line in read_object:
        print(line)
```

```python
# writing multiple lines data in a csv file
import csv
line1 = [["Name","Age","Gender"],["Ahmed",24,"Male"]]

# python 3 by defaulte create an empty line after writing each line, so we can set newline="", not no new line will be created after writing each line

# w is for wrting (overwriting on previous content), "a" is for appending below the previous content
# for python 2 remove newline and instead of just "w" use "wb" now no new line will be created
with open("mycsv.csv","w",newline="") as csv_file:
    # now file has become write object we can wrtie row by row data in it
    file = csv.writer(csv_file)
    file.writerows(line1)
    
```

```python
# reading the above created file
with open("mycsv.csv") as read_csv:
    read_object = csv.reader(read_csv)
    for line in read_object:
        print(line)
```
