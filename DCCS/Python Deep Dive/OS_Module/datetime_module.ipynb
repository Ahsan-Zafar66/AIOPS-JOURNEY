```python
# importing datetime module from datetime library
from datetime import datetime as dt
```

```python
today_date = dt.now()
today_date.year
```

```python
# strptime: Parses a string representation of a date and time into a datetime object.
# strptime, takes a string date and convert it into desire format
# string time is 31st december 2 pm 20 minutes and 40 seconds
mydate = "31-12-2023 14:30:40"
desired_format = "%d-%m-%Y %H:%M:%S"
conversion = dt.strptime(mydate,desired_format)
print(conversion)
print(type(conversion))
```

```python
# strftime: Converts a datetime object into a string representation based on a specified format
mydate2 = dt(2023,12,31, 15,30,40)
desired_format2 = "%Y-%m-%d %H:%M:%S"
conversion2 = dt.strftime(mydate2,desired_format2)
print(conversion2)
print(type(conversion2))
```

```python
# get the difference between 12,31,2023 and 12,31,2022 in seconds
first_date = dt.strptime("2:30:30","%H:%M:%S")
second_date = dt.strptime("12:01:30","%H:%M:%S")
difference = first_date - second_date
difference
```

```python
# get the difference between 12,31,2023 and 12,31,2022 in seconds
first_date = dt.strptime("2023-12-31 3:30:40","%Y-%m-%d %H:%M:%S")
second_date = dt.strptime("2022-12-31 4:30:50","%Y-%m-%d %H:%M:%S")
difference = first_date - second_date
print(difference)
print("Total seconds",difference.total_seconds())
```

```python
# getting the file created time and converting it into normal time
# listing all the files in present directory
os.listdir()

# getting the time in seconds for the file
file_time = os.path.getctime("loops.ipynb")
file_normal_time = dt.fromtimestamp(file_time)
print(f"The file is created on {file_normal_time}")
print("Current time",dt.now())
```

```python
# making a function that tells created time of the enteed file
print("Following are the files in the current working directory")

# listing the files in the directory
print(os.listdir())

# asking user to input file name
enter_file = input("Enter the name of file you want to see the created time!")

# iterating over the roots, directories and files of the currect working directory
for roots,dirs,files in os.walk(os.getcwd()):
    
    # iterating only on file present in each directory
    for file in files:

        # checking if the entered file is present in directory
        if enter_file == file:

            # computing the time in seconds of the file by combining its full path
            get_ctime = os.path.getctime(os.path.join(roots,file))
            
            # converting time from seconds into normal time
            normal_time = dt.fromtimestamp(get_ctime)

            # getting full path of file
            file_full_path = os.path.join(roots,file)
            
            # converting the normal time into desired format
            converted_time = normal_time.strftime("%A(%d) %B, %Y %I:%M:%S %p")

            # making the print statements to show the file entered, file full path and time when file is created into the mentioned format
            print(f"Entered file is {enter_file}")
            print(f"Complete file path is {file_full_path}")
            print(f"Created time of file is {converted_time}")
```
