```python
import platform
```

```python
import os
```

```python
import subprocess
```

```python
# different platforms has different commands, such as linux:ifconfig
# windows: ipconfig
# How to deal?
if platform.system() == "Windows":
    print(subprocess.run(["ipconfig"],capture_output=True,text=True))
elif platform.system() == "Linux":
    print(os.system("ifconfig"))
```
