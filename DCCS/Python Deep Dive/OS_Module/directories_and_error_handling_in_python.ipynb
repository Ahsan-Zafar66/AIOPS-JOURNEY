```python
try:
  with open("/content/module_notes.txt","r") as file:
    for line in file.readlines():
      print(line)
except FileNotFoundError:
  print("file not found")

else:
  print("The exception is handeled")
```

```python
# else statement is executed when the code in try block smoothly executed
# finally statement is executed whether the code in try block runs or not
try:
  print(10/0)
except FileNotFoundError:
  print("file not found")
except ZeroDivisionError:
  print("Zero division error, you are dividing by zero")
else:
  print("The exception is handeled")
finally:
  print("Finally statement runs when error occurs or not. ")
```
