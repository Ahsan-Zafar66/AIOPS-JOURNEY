#### Conversion of text into speech

```python
from gtts import gTTS
from IPython.display import Audio, display

def text_to_speech(text, lang='en', slow=False):
    
    # Creating a gTTS object, it takes the text in the mentioned language and convert it into audio
    tts = gTTS(text=text, lang=lang, slow=slow)

    # Saving the converted audio to a file
    audio_file = "output.mp3"
    tts.save(audio_file)

    # Displaying an audio player in the notebook
    display(Audio(audio_file, autoplay=True))

# Examp usage
instructions = "Hello! Can you please confirm if this is your correct address?"
text_to_speech(instructions)
```

#### conversion of speech into text

```python
import speech_recognition as sr

# defining a function
def speech_to_text():

    # establisshing a recognizer
    recognizer = sr.Recognizer()

    # opening the microphone 
    with sr.Microphone() as source:
        print("Listening...")
        # listening the audio
        audio = recognizer.listen(source)
    
    try:
        # recogniszing the text in the audio
        text = recognizer.recognize_google(audio)
        print(f"Recognized text: {text}")
        return text
    # if some error occured defining the exception for it
    except sr.UnknownValueError:
        print("Sorry, I could not understand the audio.")
        return None
```

```python
instructions = "Hello! Can you please confirm if this is your correct address for dr Jhon?"
text_to_speech(instructions)
output = speech_to_text()
```
