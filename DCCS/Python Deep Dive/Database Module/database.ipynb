#### entering data from dynamic sources like json file and others

```python
 # establishing the connection with the server using it's ip and giving username and password and telling the database to connect
mydb = pymysql.connect(host="192.168.43.25",user="mysql_user2",password="Test?1234?",database="alnafi")

# initiating the cursor
cur = mydb.cursor()


# data to enter in database
data = ("usman","ashraf","pharmacist","usamn11","usman1234",2000,"2024-10-01","usman@gmail.cpm")
# defining my command, yyou have to right %s the number of times eqal to number of columns, I have 8 columns and i have written %s 8 times
# the mysql by default takes the data in stirng and later convert them into their respective data types
sql_command = """ insert into trainer_detail (fname,lname,designation,username,password,salary,join_date,email) values (%s,%s,%s,%s,%s,%s,%s,%s)"""

# executing the command and passing the data
cur.execute(sql_command,data)

# commiting change from windows into linus
mydb.commit()


# defining my command
sql_command = """ select * from trainer_detail
"""

# executing the command
cur.execute(sql_command)

# bringing all the results
result = cur.fetchall()

# printing the results
print(result)

# closing the connection
mydb.close()
```
