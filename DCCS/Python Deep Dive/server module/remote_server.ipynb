#### Establishing communication without password
- go to linux machine and run "ssh-keygen" command
- now go the folder where key is created and run command "ssh-copy-id machine_ip_address" like "ssh-copy-id 192.168.43.63".
- Now, your machine has been added in the tokens.
- Now it will not ask you password.
- Now, copy that private key and use it anywhere to log into your 192.168.43.63 machine from anywhere
