```python
import paramiko
```

#### Downloading files from linux machine into windows machine

```python
# machine ip address which you want to access remotely
machine_ip = "192.168.43.211"

# hostname
machine_host = "ahsan"

# hostname apssword
machine_host_pass = "AhsanHacker?1122?"

# establishing the connection using paramiko
client = paramiko.SSHClient()

# setting the host key
client.set_missing_host_key_policy(paramiko.AutoAddPolicy())

# passing the machine ip, username and password
client.connect(hostname=machine_ip,username=machine_host,password=machine_host_pass)

# establishing sftp
mysftp = client.open_sftp()


# downloading a file from linux machine
mysftp.get("/home/ahsan/remote.py","remote.py")


# closing the client
client.close()

```

#### Sending files from windows machine to linux machine

```python
# machine ip address which you want to access remotely
machine_ip = "192.168.43.211"

# hostname
machine_host = "ahsan"

# hostname apssword
machine_host_pass = "AhsanHacker?1122?"

# establishing the connection using paramiko
client = paramiko.SSHClient()

# setting the host key
client.set_missing_host_key_policy(paramiko.AutoAddPolicy())

# passing the machine ip, username and password
client.connect(hostname=machine_ip,username=machine_host,password=machine_host_pass)

# establishing sftp
mysftp = client.open_sftp()

# changing working directoy in remote mahcine
mysftp.chdir("/home/ahsan/Downloads")

# listing files in the remote directory
print(mysftp.listdir())

# sending a file from windows to linux, sending thefile.py to linux in directory /home/ahsan/
mysftp.put("thefile.py","/home/ahsan/thefile.py")

# getting details about the file
print(mysftp.stat("/home/ahsan/thefile.py"))

# closing the client
client.close()

```
