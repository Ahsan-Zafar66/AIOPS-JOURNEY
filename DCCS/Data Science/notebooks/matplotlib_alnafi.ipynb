```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# SOME IMPORTANT COMMANDS
%matplotlib inline # to display plots in jupyter notebook

plt.tight_layout() # to prevent overlapping of x and y axis labels in subplots
```

```python
df = sns.load_dataset("penguins")
```

```python
# checking data
df.head()
```

```python
# making subplots
fig,axes = plt.subplots(1,2)
sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df,ax=axes[0])
axes[0].set_title("First plot")

sns.lineplot(x="bill_length_mm",y="bill_depth_mm",data=df,ax=axes[1])
axes[1].set_title("Second plot")
plt.show()
```

```python
# another ways of making subplots separately
# defining subplots one by one
plt.subplot(1,2,1)
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df)
plt.title("First plot")

# defining subplots one by one
plt.subplot(1,2,2)
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=df)
plt.title("Second plot")
plt.tight_layout() # prevent overlapping of x and y axis labels in subplots
plt.show()
```

```python
# making subplots one inside another

# defining figure
fig = plt.figure()

# defining subplot 1

# defining dimension of first subplot
axes1 = fig.add_axes([0.1,0.1,0.8,0.8])
axes1.set_title("First plot")
sns.scatterplot(x="bill_length_mm",y="body_mass_g",data=df,ax=axes1)

# making second subplot inside the first plot

# defining dimension of second subplot
axes2 = fig.add_axes([0.2,0.5,0.4,0.3])
axes2.set_title("Second plot")
sns.scatterplot(x="bill_length_mm",y="body_mass_g",data=df,ax=axes2)

fig.show()

# dpi controls dimensions larger value big picture smaller value small picture
fig.savefig("firssample.png",dpi=300)
fig.savefig("secondsample.png",dpi=100)
```
