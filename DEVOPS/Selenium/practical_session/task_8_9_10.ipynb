```python
# Handling different website in same code.
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
# defining broswer
driver = webdriver.Chrome()

# opening website
driver.get("https://alnafi.com/courses/course/sysops-administrator")

time.sleep(5)

# opnening another tab as separate  course
driver.execute_script("window.open('https://alnafi.com/courses/course/softskills-development','new window')")


# maximizing window
driver.maximize_window()

# getting the tabs
wins = driver.window_handles
time.sleep(2)



# swithcing to first tab
driver.switch_to.window(wins[0])
print(driver.title)

time.sleep(2)

# finidng the course name
course_title = driver.find_element(By.XPATH,'//*[@id="__nuxt"]/div/div[4]/div/div[2]/div/div[1]/h1').text

time.sleep(2)
# getting the fees of the course
course_fee = driver.find_element(By.XPATH,'//*[@id="__nuxt"]/div/div[4]/div/div[2]/div/div[2]/div/div[2]/div[1]/p[1]').text

print("Course name",course_title,"fee is",course_fee)


time.sleep(5)

# switching to second tab
driver.switch_to.window(wins[1])
# saving windows

time.sleep(5)

# finidng the course
course_title_2 = driver.find_element(By.XPATH,'//*[@id="__nuxt"]/div/div[4]/div/div[2]/div/div[1]/h1').text

time.sleep(5)

# getting the fees of the course
course_fee_2 = driver.find_element(By.XPATH,'//*[@id="__nuxt"]/div/div[4]/div/div[2]/div/div[2]/div/div[2]/div[1]/p[1]').text
print("=" * 40)

print("Course name 2",course_title_2,"fee 2is",course_fee_2)



time.sleep(5)
driver.quit()
```

```python
import pandas as pd
```

```python
courses = [course_title,course_title_2]
fees = [course_fee,course_title_2]

# creating pandas adataframe
df = pd.DataFrame({"courses":courses,"Fee":fees})
df.to_csv("data.csv")
```
