### Difference in LinkText and PartialLinkText
- Suppose you want to click on Log in button,
- linktext: driver.find_element(By.LinkText,"Log in")
- PartialLinkText: driver.find_element(By.PartialLinkText,"in")

### Implicit wait
- Sometimes what happens when we use time.sleep(5), website depending upon internet or other performance issues takes
more time to load, like 10 seconds, and next process stops because website took more time than mentioned.
- So, to deal with suh inconsistancies we use implicit wait, for example I set 20 seconds implicit wait, if website
responded in 5 seconds we will fo to next process, if it took 15 seconds it goes to next step. It will not wait 20 seconds for each step, it will wait upto maximum of 20 seconds for each process, if process is completed in less 
time it moves forwatd to next process, it is not hard coded to wait 20 seconds even after the process is completed like
time.sleep(20), it waits for 20 seconds even the process is done in 2 seconds.
- So, implicit wait connects the website response and our code.

# Disadvantage of implicit wait
- Sometimes process move so fast that some error occur in some processes, in that case identify the element where
error is occuring and the use time.sleep() function there to hold for some time before or after that process

### Explicitly Wait
- time.sleep() is hard coded and suppose if there are 20 steps or processes in your script and you add 5 second sleep so it will wait 1 min and 40 second even the process can be completed in shorter time and where steps are more than 20 then you have to wait alot to make sure process is completed. So, it is not a good choice.
- implicit wait solve the hard wait by providing soft wait that it will complete processes in speed and wait for a mentioned period of time for each process, but sometime due to internet problems it skips does not wait and process is stuck.
- So, there is explicit wait. that you define, you define time and call it before element that is causing trouble. It will wait for the mentioned period of time before executing the element next to it, and if process is complete din shorter time it will complete it in shorter time like implicit wait.

```python
from selenium import webdriver

# importing webdriver wait which is for explicit wait
from selenium.webdriver.support.wait import WebDriverWait


# importing condition to add in explicit wait
from selenium.webdriver.support import expected_conditions


# import the By element
from selenium.webdriver.common.by import By

# importing time
import time
```
