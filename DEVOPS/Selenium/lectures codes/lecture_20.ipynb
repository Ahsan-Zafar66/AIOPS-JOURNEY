### Lecture 19 XPATH CONTINUED
- XPATH Axes:
    1) Ancestor (parent, grandparent everything of an element)
    2) Parent 
    3) Sibling
    4) Child
    5) Descendant (child, subchild of an element)

### Example of Ancestor and Parent
- I am selecting a partcular element which is under form tag, //form[@class="w-full space-y-2 md:mt-8 mt-4"]/ancestor::div,  it will show me the ancestors of this tag.
- To see the parents of element //form[@class="w-full space-y-2 md:mt-8 mt-4"], we can write /parent::div then the complete syntax will become //form[@class="w-full space-y-2 md:mt-8 mt-4"]/parent::div.
- To see the child of element: //form[@class="w-full space-y-2 md:mt-8 mt-4"], you can write child like /child::div then complete synxtax will become //form[@class="w-full space-y-2 md:mt-8 mt-4"]/child::div
- To see the descendant of element: //form[@class="w-full space-y-2 md:mt-8 mt-4"], you can write descendant like /descendant::div then complete synxtax will become //form[@class="w-full space-y-2 md:mt-8 mt-4"]/descendant::div


# Difference Between Preceedng and following siblings:
- Preceeding sibling means before the element where you are standing, if there are siblings before the element where you are standing show them.
- For example: class="w-full min-h-screen flex items-center justify-between lg:h-screen"/preceeding-sibling::div

- Following sibling means after the element where you are standing, if there are siblings after the element where you are standing show them.
- For example: class="w-full min-h-screen flex items-center justify-between lg:h-screen"/following-sibling::div
