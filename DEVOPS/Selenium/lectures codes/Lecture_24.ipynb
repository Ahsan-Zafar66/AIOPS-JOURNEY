### lecture 24 Handling iframe
- iframe means that webpage where we are standing using content of nother webpage, such element on a webpage are under iframe tag.
- we can swtch to an iframe using it's id, name, index and web elements (xpath and etc)


```python
# importing libraries
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.alert import Alert
import time
```

```python
# starting browser
driver = webdriver.Chrome()


# opening website
driver.get("https://www.rediff.com/")

driver.implicitly_wait(10)

# maximizing the window
driver.maximize_window()

# now as another webpage data is being displayed in this webpage, I will switch to that webpage and come to present one
driver.switch_to.frame("moneyiframe")

# now locating the element of this webpage
myvalue = driver.find_element(By.ID,"nseindex")

print(myvalue.text)

myvalue2 = driver.find_element(By.ID,"NseChange")
print(myvalue2.text)

time.sleep(2)

driver.quit()
```

### Counting number of iframe sin a webpage

```python
driver = webdriver.Chrome()

# # opening website
driver.get("https://www.rediff.com/")

driver.implicitly_wait(10)

# # maximizing the window
driver.maximize_window()

iframes = driver.find_elements(By.TAG_NAME,"iframe")

print(len(iframes))


driver.quit()
```
