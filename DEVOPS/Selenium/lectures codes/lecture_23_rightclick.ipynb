### How to right click on a button


```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
import time
```

```python
# opening the browser
driver = webdriver.Chrome()

driver.implicitly_wait(10)

# opening the website
driver.get("https://www.selenium.dev/documentation/")

# download element
download = driver.find_element(By.XPATH,'//*[@id="main_navbar"]/ul/li[2]/a/span')

# action chain
action = ActionChains(driver)

# using action chain 
action.context_click(download).perform()

time.sleep(5)

driver.quit()
```

### How to double click

```python
# opening the browser
driver = webdriver.Chrome()

driver.implicitly_wait(10)

# opening the website
driver.get("https://www.selenium.dev/documentation/")

# download element
download = driver.find_element(By.XPATH,'//*[@id="main_navbar"]/ul/li[2]/a/span')

# action chain
action = ActionChains(driver)

# using action chain 
action.double_click(download).perform()

time.sleep(5)

driver.quit()
```

### How to handle alert statement
- To handle alert statement, you need to look after which step it occurs or by clicking at which button it occurs.

```python
from selenium.webdriver.common.alert import Alert
```

```python
# starting chrome browser
driver = webdriver.Chrome()

# opening the website
driver.get('https://mail.rediff.com/cgi-bin/login.cgi')

# finding the sign in button and clicking on it to create alert
driver.find_element(By.XPATH,'/html/body/div[1]/div[1]/div[1]/div[2]/form/div[1]/div[2]/div[2]/div[2]/div/input[2]').click()

time.sleep(5)

# inputing alert from driver into Alert function
alert = Alert(driver)

# printing the alert statement
print(alert.text)

# accepting the alert
alert.accept()

time.sleep(5)

driver.quit()
```

### How to show text of something on website

```python
# starting browser
driver = webdriver.Chrome()

# opening website
driver.get("https://www.rediff.com/")

# maximizing window
driver.maximize_window()

# finding element
myelement = driver.find_element(By.XPATH,'/html/body/div[12]/div[2]/p/a')

# printing the text
print(myelement.text)

# quiting website
driver.quit()
```
