```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
import time
```

```python
# starting the browser
driver = webdriver.Chrome()

# opening the website
driver.get("https://www.flipkart.com/")

# maximizing window after opening the website
driver.maximize_window()

# adding implicit wait
driver.implicitly_wait(10)

# finding menu in flipkart and saving it in a variable
menu = driver.find_element(By.XPATH,'//*[@id="container"]/div/div[1]/div/div/div/div/div[1]/div/div/div/div[2]/div[1]/div/div[1]/div/div/div/div/div[1]/div[2]/div/div/span/span[1]')

# we always pass the driver in action chain
action = ActionChains(driver) 

# now I want to move to the element which is my electronics menu saved in variable menu
action.move_to_element(menu).perform()

time.sleep(2)


# now i will click on my element which is Electronics GST Store inside electronics hover over menu
driver.find_element(By.LINK_TEXT,"Electronics GST Store").click()


# sleeping for 10 seconds
time.sleep(5)

# quiting driver
driver.quit()
```

### Dealing with drag and drop buttons on website
- First find the drag and drop element
- After finding drag and drop element make changes in bar and see how value changes in inspect section

```python
# we can directly open a page y adding its link
driver = webdriver.Chrome()

# opening page directly in flipkart of wired phones
driver.get('https://www.flipkart.com/audio-video/headphones/pr?sid=0pm%2Cfcn&otracker=categorytree&p%5B%5D=facets.connectivity%255B%255D%3DWired&fm=neo%2Fmerchandising&iid=M_9454f6b7-3721-4176-88b4-b9aab53ff187_1_372UD5BXDFYS_MC.LSKPCXUOS6SB&otracker=hp_rich_navigation_3_1.navigationCard.RICH_NAVIGATION_Electronics~Audio~Wired%2BHeadphones_LSKPCXUOS6SB&otracker1=hp_rich_navigation_PINNED_neo%2Fmerchandising_NA_NAV_EXPANDABLE_navigationCard_cc_3_L2_view-all&cid=LSKPCXUOS6SB')

# maximizing the window
driver.maximize_window()

# finding the drag and drop bar
drag_drop = driver.find_element(By.XPATH,'//*[@id="container"]/div/div[3]/div/div[1]/div/div[1]/div/section[2]/div[3]/div[1]/div[2]/div')

#drag_drop = driver.find_element(By.XPATH,'//*[@id="container"]/div/div[3]/div/div[1]/div/div[1]/div/section[2]/div[3]/div[1]/div[2]/div')



# now using action chain
action_chain = ActionChains(driver)

# now performing action on the element which is drag_drop
# drag_drop_offset takes 3 aguments, first is the element path, second x-scale value and third y-scale value
# i am dropping the value of x-scale, when you call inspect on drag and drop change value of slider
# there will be decrease in px value note it down  and manipulate the slider according to it
action_chain.drag_and_drop_by_offset(drag_drop,-150,0).perform()

# adding sleep time
time.sleep(10)

# quiting the window
driver.quit()
```
