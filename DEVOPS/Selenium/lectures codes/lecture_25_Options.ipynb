```python
# importing libraries
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.alert import Alert
import time
```

```python
# starting chrome options
chrome_option = webdriver.ChromeOptions()

# setting the headless option, so whatever I do happens in the background
chrome_option.add_argument('--headless')

# opening the browser
driver = webdriver.Chrome(options=chrome_option)

# opening the website
driver.get("https://alnafi.com/")

time.sleep(5)
# printing the title of the website
print(driver.title)

time.sleep(5)

driver.quit()
```

```python
from selenium import webdriver

# Set up Chrome options
chrome_option = webdriver.ChromeOptions()


# Add headless option to run without opening the browser
chrome_option.add_argument('--headless')

# Initialize the Chrome WebDriver with options
driver = webdriver.Chrome(options=chrome_option)

driver.implicitly_wait(20)

# Open YouTube website
driver.get("https://alnafi.com/")

time.sleep(5)

# Print the title of the website
print("Page Title:", driver.title)

# Close the WebDriver
driver.quit()
```
