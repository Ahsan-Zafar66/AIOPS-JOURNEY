### CSS Selector
- Following is the syntax of css selector
    - #id. For example highlighting google sign in email section. email section id in the form of css selector is #identifierId
    - tagname#id. For example the email section in google sign in has id identifierId. Now, it has tagname input. So, css selctor in this format will become input#identifierId.
    - tagname[attribute="value"]. For example input[id=identifierId]

### Scenario where attribute changes continously like in banking websites
- So, in scenario where attribute which is id, name and any other thing changes continously what to do there.
- So, suppose id of input section in a website is username, but next day it becomes userinput.
- So, we can search for a pattern like, input[id^=user], ^ means id which start with user, or input[id$=user] end with user.
- What to do in scenario where 123user456, how you will get user? user is in between, so pattern to track in between thing is input[id*='user/']



### Parent to Child Navigation
- TO navigate from parent to child, you first need parent tag and classname and then you can navigate or see what tags this parent has.
- For example in google sign in page, I navigated from the parent using it's tag and classname to it's child.
- div.S7xv8.LZgQXe >div > div > div > div > span
- div.S7xv8.LZgQXe, the div is parent tag and after . is it's class. and then using > div i am goinf to child then that tag child and so on, where > span is used it means tag name is no longer div, so tag can get change 
- if you get stuck after span in, "div.S7xv8.LZgQXe >div > div > div > div > span", you can use > * to check what tags are available after span like: "div.S7xv8.LZgQXe >div > div > div > div > span > *"

- if you want to see the first child in a tag then you can do this, div.S7xv8.LZgQXe > div > *:first_child, if there are a lot of childs and you want to go to the 4th child then you can do this, div.S7xv8.LZgQXe > div > *:nth-child(4) 



```python
# importing the libraries
import selenium
from selenium import webdriver

from selenium.webdriver.common.by import By
import time
# importing select
from selenium.webdriver.support.select import Select
```

```python
# CSS Selector: Cascading style sheets
# this is the css selector for the wikipedia input section #searchInput

driver = webdriver.Chrome()

driver.get("https://www.wikipedia.org/")

driver.find_element(By.CSS_SELECTOR,"#searchInput").send_keys("Hi! this is by css selector")

time.sleep(5)
driver.quit()
```
