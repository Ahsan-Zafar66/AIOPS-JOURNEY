### Multiple Check box by automation
- Checkbox XPATH :-
    1) 5 --> //*[@id="orders-grid"]/tbody/tr[1]/td[1]/input

    2) 4 --> //*[@id="orders-grid"]/tbody/tr[2]/td[1]/input

    3) 3 --> //*[@id="orders-grid"]/tbody/tr[3]/td[1]/input

    4) 2. --> //*[@id="orders-grid"]/tbody/tr[4]/td[1]/input

    5) 1 --> //*[@id="orders-grid"]/tbody/tr[5]/td[1]/input

    - driver.find_element(By.XPATH,'//*[@id="orders-grid"]/tbody/'+str(i)+'/td[1]/input').click()

```python
# starting chrom
driver = webdriver.Chrome()

# opening website
driver.get("https://admin-demo.nopcommerce.com/Admin/Order/List")

# maximizing window
driver.maximize_window()

# finding the email element
my_email = driver.find_element(By.ID,"Email")
time.sleep(2)

# clearing already filled value
my_email.clear()
time.sleep(2)

# entering my own value
my_email.send_keys('admin@yourstore.com')
time.sleep(2)

# finding  password element
my_pass = driver.find_element(By.ID,"Password")

# clearing already filled password
my_pass.clear()

# entering my own password
my_pass.send_keys('admin')

# finding the login button and ticking it
driver.find_element(By.XPATH,'/html/body/div[6]/div/div/div/div/div[2]/div[1]/div/form/div[3]/button').click()

time.sleep(3)

# making a function that takes in which element to look for and it's value, if element present True else False
def is_elements_avail(att,abd):
    if len(driver.find_elements(by=att,value=abd)) == 0:
        return False
    else:
       return True

# defining value of i
i= 1

# starting a while loop where I am entering the checkbox xpath and sending it in the function to check if it
# is present, if present then click the check box
while is_elements_avail(By.XPATH,'//*[@id="orders-grid"]/tbody/tr['+str(i)+']/td[1]/input'):
    driver.find_element(By.XPATH,'//*[@id="orders-grid"]/tbody/tr['+str(i)+']/td[1]/input').click()
    time.sleep(1)
    
    # increasin value of i after each checkbox as they have pattern in them
    i = i + 1

# exiting the browser
driver.quit()
```

### Method 2

```python
# Method 2 :-
# starting chrome
driver = webdriver.Chrome()

# opening the webpage
driver.get("https://admin-demo.nopcommerce.com/Admin/Order/List")

# maximizing the window
driver.maximize_window()

# finding email,clearing it and entering own value
my_email = driver.find_element(By.ID,"Email")
time.sleep(2)
my_email.clear()
time.sleep(2)
my_email.send_keys('admin@yourstore.com')
time.sleep(2)

# finding password clearing it and entering own value
my_pass = driver.find_element(By.ID,"Password")
my_pass.clear()
my_pass.send_keys('admin')

# finf the login button and click it
driver.find_element(By.XPATH,'/html/body/div[6]/div/div/div/div/div[2]/div[1]/div/form/div[3]/button').click()

time.sleep(3)

# fetching the check box group, it contain all the checkboxes on the open page
my_checkboxes = driver.find_elements(By.CLASS_NAME,"checkboxGroups")

# now running a for loop on the checkboxes and clicking all of them
for i in my_checkboxes:
    i.click()
time.sleep(0.2)

# printing total number of checkboxes
print("Total checkbox value is : ",len(driver.find_elements(By.CLASS_NAME,"checkboxGroups")))
driver.quit()

#############################
```
