# Lecture_14 DROPDOWN MENUS

```python
# importing the libraries
import selenium
from selenium import webdriver

from selenium.webdriver.common.by import By
import time
# importing select
from selenium.webdriver.support.select import Select
```

```python
# initiating browser chrom
driver = webdriver.Chrome()

# fetching the website
driver.get("https://www.saucedemo.com/")

# maximizing the window
driver.maximize_window()


time.sleep(3)
# finding the username and entering the username
driver.find_element(By.ID,"user-name").send_keys("standard_user")
time.sleep(3)

# finding the password and entering the password
driver.find_element(By.ID,"password").send_keys("secret_sauce")
time.sleep(3)

# clicking the login or sign in button
driver.find_element(By.ID,"login-button").click()
time.sleep(3)

# finding the select menu and entering its option
driver.find_element(By.CLASS_NAME,"product_sort_container").send_keys("Name (Z to A)")
time.sleep(5)

driver.quit()
```

### Task 1 Count How Many Languages Are Available in Wikipedia

```python
driver = webdriver.Chrome()

driver.get("https://www.wikipedia.org/")

driver.maximize_window()

time.sleep(4)

# i am finding the dropdown menu and saving it in a variable
mymenu = driver.find_element(By.ID,"searchLanguage")

# I am inputing my dropdown menu into the select statement
# when we use select for dropdown menus it provides us witha lot of options
myselection = Select(mymenu)

# I am saving all the available options in dropdown menu in a variable
num_option =  myselection.options


# printing the languages and their values
# we first fetch the text of the element and then the attribute which is value stored under lang tag
for i in num_option:
    print("Name of language is",i.text,"Value of language is",i.get_attribute("lang"))

# now I will count the options
print("The number of languages available in wikipedia are:",len(num_option))

time.sleep(4)

driver.quit()
```

### Task 2 Find all the links in a webpage
- as links are stored as hyperlinks, so hyperlinks are stored inside  tag (a), so you need to enter in this tag and count the elements present in it
- we use find_elements instead of find_element


```python
# initiating chrome
driver = webdriver.Chrome()

# fetching the website
driver.get("https://alnafi.com/")

#driver.get("https://www.wikipedia.org/")
# maximizing the window
driver.maximize_window()

time.sleep(3)

# entering the <a> tag, where links are stored
num_links = driver.find_elements(By.TAG_NAME,"a")

# iterating over the elements of links
for link in num_links:

    # getting the link text and actual link, links are mostly stored under href tag
    print("Link name is",link.text,"actual link is",link.get_attribute("href"))

print("The number of links in the alnafi welcome page are",len(num_links))

time.sleep(5)
driver.quit()
```

```python
# above cell is for all the links in a webpage
# But, what if I want only a section and links available in that section?
# initiating chrome
driver = webdriver.Chrome()

# fetching the website
driver.get("https://alnafi.com/")

#driver.get("https://www.wikipedia.org/")
# maximizing the window
driver.maximize_window()

time.sleep(15)

# selecting course section at the bottom of page using its XPATH
# so in following line we are selecting course section
course_section = driver.find_element(By.XPATH,'//*[@id="footer-section"]/div/div[2]/div[1]/div[2]')

# now we will look for a tag in this course section
# now we can access all the links present inside the course section
num_links = course_section.find_elements(By.TAG_NAME,"a")

# if we want to accessa particular item or element we can access like that
print("Second  Element")
print(num_links.__getitem__(1).text)
#print(course_section.find_elements(By.TAG_NAME,"a").__getitem__(1).text)
print("=" * 50)

# OR you can append in a list and access it later
my_links = []

# iterating over the elements of links
for link in num_links:

    # getting the link text and actual link, links are mostly stored under href tag
    print("Link name is",link.text,"actual link is",link.get_attribute("href"))
    my_links.append(link.text)
print("The number of links in the alnafi course section at bottom of page are",len(num_links))



print("Second element by list method",my_links[1])
time.sleep(5)
driver.quit()
```
