```python
# importing libraries
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.alert import Alert
import time
```

```python
# calling a function of html, we can click sign in but I am calling function so there is no need to click sign in
# previously we clicked sign in button by finding it's element and get the warning, now we are calling the function to get the warnng

# starting chrome
driver = webdriver.Chrome()

# opening website
driver.get('https://mail.rediff.com/cgi-bin/login.cgi')

# waiting implicitly
driver.implicitly_wait(10)

# maximizing wnindow
driver.maximize_window()

# I am running a function which is defined in webpage html
driver.execute_script("checkLogin()")

time.sleep(5)

driver.quit()
```

### Taking screenshot

```python
# starting browser
driver = webdriver.Chrome()

# opening website
driver.get("https://mail.rediff.com/cgi-bin/login.cgi")

time.sleep(3)

# taking screenshot
driver.save_screenshot("myfirstscreenshot2.png")

time.sleep(1)

driver.quit()
```

```python
# Create a directory and you can give your path to save the png file

# starting browser
driver = webdriver.Chrome()

# opening website
driver.get("https://mail.rediff.com/cgi-bin/login.cgi")

time.sleep(3)

# taking screenshot and saving in a directory name myscreenshots
driver.get_screenshot_as_file("./myscreenshots/myfirstscreenshot2.png")

time.sleep(1)

driver.quit()
```

### How to get screenshot of a particular in a webpage

```python
# starting browser
driver = webdriver.Chrome()

# opening website
driver.get("https://alnafi.com/")

time.sleep(3)

# finding the area and saving in a variable
particular_area = driver.find_element(By.XPATH,'//*[@id="footer-section"]/div/div[2]/div[2]')

time.sleep(2)

# taking screenshot of the particular area
particular_area.screenshot("address.png")

# quiting the webpage
driver.quit()
```
