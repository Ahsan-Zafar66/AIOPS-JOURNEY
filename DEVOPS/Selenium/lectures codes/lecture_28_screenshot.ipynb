### How to highlight something in a webpage

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
from datetime import date,datetime
```

```python
# when we want to highlight something, we first to need to locate that element and then right in htmly language

# opening chrome
driver = webdriver.Chrome()

# fetching the website
driver.get("https://github.com/login")

# maximizing the window
driver.maximize_window()

# now we are locating sign in box
signin = driver.find_element(By.XPATH,'//*[@id="login_field"]')

# now we are executing the script to make the box purple and borders are yellow
driver.execute_script("arguments[0].setAttribute('style','background: purple; border:2px solid yellow;');",signin)
driver.save_screenshot("github_colored.png")

time.sleep(2)

driver.quit()

```

### Creating  a screenshot with label or date

```python
# getting current date
day = date.today()

# getting the current time
date_time=datetime.now()

# converting the day format to a the format we like
date_format=day.strftime("%B_%d_%Y")

# converting time into format we like
time_format=date_time.strftime("%I_%M_%p")

# name of screenshot with which it will be saved
myfile = "Screenshot__"+date_format+"_"+time_format+".png"

# starting chrome browser
driver = webdriver.Chrome()

# fetching the website
driver.get('https://github.com/login')

# maximizing window
driver.maximize_window()

# waiting implicitly
driver.implicitly_wait(10)

time.sleep(1)

# now we are locating sign in box
signin = driver.find_element(By.XPATH,'//*[@id="login_field"]')

# now we are executing the script to make the box purple and borders are yellow
driver.execute_script("arguments[0].setAttribute('style','background: purple; border:2px solid yellow;');",signin)
driver.save_screenshot(myfile)

time.sleep(2)
driver.quit()
```

### Saving a file by making a function

```python
# Now we will create own time file function in selenium

driver = webdriver.Chrome()

# defining the function that takes in driver as an argument
def my_file_sc(driver):
    file_name = time.asctime().replace(":","_")+".png"
    driver.save_screenshot(file_name)

# fetching the website
driver.get('https://github.com/login')

# maximizing window
driver.maximize_window()

# waiting implicitly
driver.implicitly_wait(10)

time.sleep(1)

# now we are locating sign in box
signin = driver.find_element(By.XPATH,'//*[@id="login_field"]')

# now we are executing the script to make the box purple and borders are yellow
driver.execute_script("arguments[0].setAttribute('style','background: purple; border:2px solid yellow;');",signin)

# passing the driver in the function
my_file_sc(driver)


time.sleep(.5)
driver.quit()
```

### How to take screenshot of entire webpage or window

```python
import os

# starting chrome
driver = webdriver.Chrome()

# making a function that store the image in the current working directory
def my_file_sc(driver):
    my_path = os.getcwd() + "\\"
    file_name = my_path+ time.asctime().replace(":","_")+"full_page.png"
    driver.save_screenshot(file_name)

# fetching the website
driver.get("https://alnafi.com")

# maximizing the window
driver.maximize_window()

# implicitly waiitng
driver.implicitly_wait(10)

# sleeping time
time.sleep(5)

# fetching the widht of webpage
total_width = driver.execute_script("return document.body.offsetWidth")

# fetching the height of webpage
total_height = driver.execute_script("return document.body.offsetHeight")
print(total_height)
print(total_width)

time.sleep(3)
# setting the window size accoridng to the height and width of entire webpage
driver.set_window_size(total_width,total_height)

time.sleep(5)

my_file_sc(driver)

time.sleep(3)
driver.quit()
```
