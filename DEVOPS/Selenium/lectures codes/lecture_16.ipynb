- is_displayed(): this gives True value if the element is present, if not present then gives exception error. is_displayed always work with hidden elements


```python
# importing the libraries
import selenium
from selenium import webdriver

from selenium.webdriver.common.by import By
import time
# importing select
from selenium.webdriver.support.select import Select
```

```python
# above cell is for all the links in a webpage
# But, what if I want only a section and links available in that section?
# initiating chrome
driver = webdriver.Chrome()

# fetching the website
driver.get("https://alnafi.com/")

#driver.get("https://www.wikipedia.org/")
# maximizing the window
driver.maximize_window()

time.sleep(5)

# using the is_displayed and checking if tag a is present in the webpage
print(driver.find_element(By.TAG_NAME,'a').is_displayed())

time.sleep(5)
driver.quit()
```
