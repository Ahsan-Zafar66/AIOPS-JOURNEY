### Lecture 19 XPATH
- XPATH in selenium is a XML PATH used for navigation through the HTML elements of the page
- It is a syntax or language for finding any elements on webpage
- XPATH always start from double slash or //
- You can make your xpath on your own, suppose sign in element in alnafi has name=email, so to locate this element you can make XPATH: //input[@name="email"]
- //tagname[@attribute="value_of_attribute"] this is the syntax of XPATH, you can take any tag, then any of it's attribute

### Types of XPATH
- There are two type sof XPATH
    - 1. Relative XPATH
    - 2. Absolute XPATH


### Join XPATH :-
- 1. Method :-
- //input[@class='whsOnd zHQkBf'][@id="identifierId"]

- OR

- 2. Method :-
- //input[@class='whsOnd zHQkBf' and @id="identifierId"]

### Finding elements by the text on the webpage
- Suppose you want to find forgot email element, you can use following command:
- //*[contains(text(),"Forgot")]  // means XPATH, * means look in all tags as no tag is specified becuase we don't kno forgot email is under which tag, [contains(text(), Forgot)] contains is a function and it takes two arguments first for what thing we are looking for in this case we are looking by text, and what is the text, as text is "Forgot email" so I only write Forgot and it captured the element.

- Finding element by attribute using contain function.
- //*[contains(@name,"identifier")] look in all tags where attribute to look is name and value of this attributr is identifier.
- Same thing using attribute id: //*[contains(@id,"identifierId")]
