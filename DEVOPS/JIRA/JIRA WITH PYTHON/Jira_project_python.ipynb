### Project Details
- In this project, we will connect to jira using an account.
- We will make connection, and use JQL to query every 2 hours to show us the unassigned and To-Do tickets in a speciifc project.
- Then, we will write emails to the reporters of issue (people who created the ticket), that their issue has been noted and we will get back to them as soon as possible.


```python
# lets import jira
from jira import JIRA
# importing smtp module
import smtplib
```

```python
# email of account through which we login
email = "use your mail"

# api token of account
api_token = "use api token of your account"

# website where we want to log in
server = "atllasian url"
```

```python
# now making a connection
jira_connection = JIRA(server=server,basic_auth=(email,api_token))
```

```python
# creating a query and running it
query_result = jira_connection.search_issues("created >= -2h AND project = LP AND assignee = empty")

# getting length of the results we got, it will tell us how many tickets are there
number_issues = len(query_result)
```

```python
# checking query results
query_result
```

```python
# creating a list to store the ticket number
tnumber = []

# creating a list to store the name of the reporter of tickets
reporters_name = []

# saving reporter of ticket email in a list, to send them mails later
# NOTE: you will need to change the status of visibility of emails by going profile visibility settings
# and set the email visibility to anyone, otherwise you will get an error
reporters_email = []


# summary of ticket
summary = []

# running a for loop on the tickets we got
for ticket in query_result:

    # extracting the 
    tnumber.append(ticket.key)

    # person who reported the issue
    reporters_name.append(ticket.fields.reporter.displayName)

    # email id of the person who reported
    reporters_email.append(ticket.fields.reporter.emailAddress)

    # appending the summary
    summary.append(ticket.fields.summary)


```

```python
# making an if else condition
if number_issues != 0:
    
    # creating a for lopp that checks how many tickets came in our query and now iterate on them one by one
    for i in range(number_issues):

        # defining the email from which mails will be sent to issue reporters
        my_mail = "use your mail"


        # mail to which message will be send, as we are iterating on tickets, so different ticket may be reported
        # by different individuals so using the index to keep track of information
        to_mail = reporters_email[i]

        # my app password set in my google account ahsanzafarqau@gmail.com
        my_pswd = "use password"

        # you need to define subject and message in capital letters
        SUBJECT=tnumber[i]
        
        # creating the message that the ticket creator will see
        MESSAGE = f"""Hi {reporters_name[i]}, thanks for reporting us the issue {tnumber[i]} that you are facing, we will get back to you as soon as possible.
        Regards,
        Alnafi Team """

        # this format is special and you need to follow it always, if you don't use \n everything will be sent as subject 
        text = f"Subject: {SUBJECT}\n\n{MESSAGE}"


        # smtp mail of google.com, as gmail works on 587 port
        connection = smtplib.SMTP("smtp.gmail.com",587)

        # starting transport layer , TLS: transport layer security
        connection.starttls()

        # logging in the account
        connection.login(user=my_mail,password=my_pswd)

        # sending mail
        connection.sendmail(from_addr=my_mail,to_addrs=to_mail,msg=text)


        # closing the connection after sending mail
        connection.close()
```
