```python
from jira import JIRA
```

```python
user = "use your mail"
api = "use your atlassian access token"
myserver = "url of server or atlassian web"
```

### NOTE
- If email and password are not working for connection with JIRA and you get an error, go to JIRA, go to settings, then go to security, click on manage API, and create an api token.
- Now, use the api token to build connection with JIRA.

### USE ONLINE JSON VIEWER 

```python
# establishing the connection with jira using username and api token
jira = JIRA(server=myserver,basic_auth=(user,api))
```

```python
# printing the fields
print(jira.fields())
```

```python
# defining an issue
ticket = "SCRUM-1"
```

```python
# getting details about the issue
myissue = jira.issue(id=ticket)
```

```python
# getting sumary about the issue
myissue.fields.summary
```

```python
myissue
```

```python
# so, we need an issue, that issue has fields and we can view information
# i am viewing the project to which this issue belongs to
myissue.fields.project.name
```

```python
# seeing the reporter of issue
myissue.fields.reporter.displayName
```

```python
# getting isssue status
myissue.fields.status.name
```

```python
# time spend on issue
print(myissue.fields.timetracking)
```

```python
# finding the comments on an issue and printing them
# here nothin appeared which shows no comment has been done on it
comment = myissue.fields.comment.comments
for i in comment:
    print("Comment is:", i.body)
    print("Id is:",i.id)
    print("Author Name is:",i.author)
    print("Time at which comment created is:",i.created)
    print("=" * 50)
```

### Show the last (update) comment

```python
comment_db = []
comment = myissue.fields.comment.comments
for i in comment:
    comment_db.append(i.body)

print("Last comment on the project is:",comment_db[-1])
```
