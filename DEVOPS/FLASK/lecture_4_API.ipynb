### HTTP Post Method
- We will use jira issue creation api for learning the post method
- http://localhost:8090/rest/api/2/issue

```python
import json
```

```python
# first we need jira issue creation api

issue_creation_api = "https://ahsanzafarqau.atlassian.net/rest/api/2/issue"

# now we need email and api token of jira account
# email of account through which we login
email = "ahsanzafarqau@gmail.com"

# api token of account
api_token = "ATATT3xFfGF0rv4nzQuXRBzq_DLrL2aoMoURxzdQgFrbGxrvLsx-Pz2nDLJjSUuFO_4PxGzrRSvIMKgozMnD9mDd-IPlvFwoqoV7F4sBfWF3F20NNLDcGKcUur3gpk-wnDj8UJWMvW3H52qvCzfSm8WyjT-E4QU2iUqtD-Y5I5SRu35Sby4x7_0=757B7A9A"


# we need the issue description now, like which issue in which project we want to create
issue_description = {
    "fields": {
        "project": {"key":"LP"},
        "issuetype": {"name":"Bug"},
        "priority":{"name":"Highest"},
        "summary": "there is a bug creation through jira issue creation api",
        "description":"bug through api",
        
    }
}


# defining headers
header = {
    "Accept": "application/json",
    "content-type": "application/json"
}

# now we need to convert over issue details in json format
json_paylod = json.dumps(issue_description)

# now we will use post method to post this on our api
response = requests.post(issue_creation_api,auth=(email,api_token),data=json_paylod,headers=header)

print(response.json())

```
